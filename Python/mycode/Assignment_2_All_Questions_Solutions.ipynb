## Q7. Three-Digit Number Analysis

Take a three-digit number, extract hundreds/tens/units digits, then calculate their sum, product, and the reverse.

```python
number = int(input("Enter a three-digit number: "))

hundreds = number // 100
tens = (number // 10) % 10
units = number % 10

digit_sum = hundreds + tens + units
digit_product = hundreds * tens * units
reverse_number = units * 100 + tens * 10 + hundreds

print("Hundreds =", hundreds)
print("Tens =", tens)
print("Units =", units)
print("Sum =", digit_sum)
print("Product =", digit_product)
print("Reverse =", reverse_number)
```

# Self-Check — Quick Answers

- **Why does `input()` return a string?** User input is read as text, so numeric input must be converted when arithmetic is required.
- **When to use `int()` and `float()`?** Use `int()` for whole-number input and `float()` for decimal input.
- **Difference between `/`, `//`, and `%`:** `/` gives division result, `//` gives floor division, and `%` gives the remainder.
- **Operator precedence:** Python evaluates higher-priority operators first; parentheses can change the order.
- **Extract digits using `//` and `%`:** `//` removes lower-place digits, while `%` extracts remainders.
- **Indexing:** Accessing an element by its position.
- **Positive vs negative indexing:** Positive indexing starts from `0` at the beginning; negative indexing starts from `-1` at the end.
- **Slicing:** Selects a range of list elements using `start:stop:step`.
- **Step in slicing:** Controls how many positions Python moves between selected elements.
- **`append()` vs `insert()`:** `append()` adds at the end; `insert()` adds at a specified index.
- **`remove()` vs `pop()`:** `remove(value)` removes a matching value; `pop(index)` removes by index, with the last item removed when no index is given.
- **Changing one list element:** Assign a new value to its index.
- **Indexes after insertion/removal:** Positions of elements after the changed position can shift.
- **Combining lists with calculations:** Use list elements and built-in functions such as `sum()`, `len()`, `min()`, and `max()`.
- **`len()`, `sum()`, `min()`, `max()`:** Number of elements, total, smallest value, and largest value respectively.
