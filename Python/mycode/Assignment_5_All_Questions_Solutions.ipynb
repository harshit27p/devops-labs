## Q12. Simple Calculator

**Question:** Take two numbers and an operator from the user. Perform +, -, *, or /. For another operator, print `Invalid operator`.

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

if operator == "+":
    print("Result:", a + b)
elif operator == "-":
    print("Result:", a - b)
elif operator == "*":
    print("Result:", a * b)
elif operator == "/":
    if b != 0:
        print("Result:", a / b)
    else:
        print("Cannot divide by zero.")
else:
    print("Invalid operator")
```

## Q16. Shopping Discount

**Question:** Take shopping amount. Apply 20% for >=10000, 10% for >=5000, 5% for >=2000, otherwise no discount. Display original amount, discount, and final amount.

```python
amount = float(input("Enter shopping amount: "))

if amount >= 10000:
    discount_rate = 0.20
elif amount >= 5000:
    discount_rate = 0.10
elif amount >= 2000:
    discount_rate = 0.05
else:
    discount_rate = 0.00

discount = amount * discount_rate
final_amount = amount - discount

print("Original amount:", amount)
print("Discount:", discount)
print("Final amount:", final_amount)
```

## Q17. Electricity Bill

**Question:** Take units consumed and calculate the bill using the assignment's rates: <=100 at ₹2/unit, 101–200 at ₹3/unit, 201–300 at ₹5/unit, and above 300 at ₹7/unit.

```python
units = float(input("Enter units consumed: "))

if units <= 100:
    rate = 2
elif units <= 200:
    rate = 3
elif units <= 300:
    rate = 5
else:
    rate = 7

bill = units * rate

print("Rate per unit: ₹", rate)
print("Total electricity bill: ₹", bill)
```

This follows the assignment literally as a **single applicable rate based on the total units**. It does not use slab-wise billing because the PDF specifies the ranges as rates per unit.

## Q19. Largest of Three Numbers

**Question:** Take three numbers, determine the largest, and appropriately handle cases where any two or all three are equal.

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))

if a == b == c:
    print("All three numbers are equal.")
elif a == b and a > c:
    print("First and second numbers are equal and largest.")
elif a == c and a > b:
    print("First and third numbers are equal and largest.")
elif b == c and b > a:
    print("Second and third numbers are equal and largest.")
elif a > b and a > c:
    print("First number is largest.")
elif b > a and b > c:
    print("Second number is largest.")
else:
    print("Third number is largest.")
```

## Q20. Movie Ticket Price

**Question:** Take age and determine the ticket price: below 5 Free, 5–12 ₹100, 13–59 ₹200, 60+ ₹120. Then ask for student ID and apply a 20% discount if eligible.

```python
age = int(input("Enter your age: "))

if age < 5:
    ticket_price = 0
elif age <= 12:
    ticket_price = 100
elif age <= 59:
    ticket_price = 200
else:
    ticket_price = 120

student_id = input("Do you have a student ID? (yes/no): ").strip().lower()

if student_id == "yes" and ticket_price > 0:
    ticket_price = ticket_price * 0.80
    print("20% student discount applied.")

print("Final ticket price: ₹", ticket_price)
```

# Quick Revision — Decision Making

### `if`
Runs a block only when a condition is true.

### `if-else`
Chooses between two alternatives.

### `if-elif-else`
Checks multiple conditions in order.

### Nested `if`
An `if` statement placed inside another `if` statement.

### Important operators used

- `>` greater than
- `<` less than
- `>=` greater than or equal to
- `<=` less than or equal to
- `==` equal to
- `!=` not equal to
- `and` both conditions must be true
- `or` at least one condition must be true
- `%` gives the remainder and is commonly used to check even/odd numbers
