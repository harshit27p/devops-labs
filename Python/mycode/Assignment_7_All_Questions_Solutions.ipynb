## Q9. *args — Calculate Sum

**Question:** Create `calculate_sum()` accepting any number of numbers using `*args` and return the total.

````python
def calculate_sum(*args):
    total = 0
    for number in args:
        total += number
    return total

print(calculate_sum(10, 20))
print(calculate_sum(10, 20, 30, 40))
````

`*args` collects any number of positional arguments into a tuple.

## Q10. *args — Find the Largest

**Question:** Create a function using `*args` that returns the largest number without using `max()`.

````python
def find_largest(*args):
    if len(args) == 0:
        return None

    largest = args[0]

    for number in args[1:]:
        if number > largest:
            largest = number

    return largest

print("Largest:", find_largest(10, 25, 7, 45, 18))
print("Largest:", find_largest(100, 50, 250, 75))
````

## Q11. **kwargs

**Question:** Create `student_details()` accepting any number of keyword arguments using `**kwargs` and print the received information.

````python
def student_details(**kwargs):
    print("Student Information:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

student_details(
    name="Rahul",
    age=22,
    course="Python",
    city="Bengaluru"
)
````

`**kwargs` collects keyword arguments into a dictionary.

## Q12. Multiple Calculations

**Question:** Create a function accepting two numbers and returning sum, difference, product, and division. Store returned values separately.

````python
def calculate(a, b):
    total = a + b
    difference = a - b
    product = a * b
    division = a / b if b != 0 else "Cannot divide by zero"
    return total, difference, product, division

sum_result, difference_result, product_result, division_result = calculate(20, 5)

print("Sum:", sum_result)
print("Difference:", difference_result)
print("Product:", product_result)
print("Division:", division_result)
````

## Q13. Return vs Print

**Question:** Create `display_square()` that prints the square and `calculate_square()` that returns the square. Call both.

````python
def display_square(number):
    print("Square printed by function:", number ** 2)

def calculate_square(number):
    return number ** 2

display_square(5)

result = calculate_square(5)
print("Square returned by function:", result)
````

`print()` displays a value. `return` sends a value back to the caller so it can be stored or reused.

## Q14. Return Multiple Values

**Question:** Create a function accepting a list and returning total, highest, lowest, and average. Store and display the results.

````python
def analyze_numbers(numbers):
    total = sum(numbers)
    highest = max(numbers)
    lowest = min(numbers)
    average = total / len(numbers)
    return total, highest, lowest, average

numbers = [10, 20, 30, 40, 50]

total, highest, lowest, average = analyze_numbers(numbers)

print("Total:", total)
print("Highest:", highest)
print("Lowest:", lowest)
print("Average:", average)
````

## Q20. Local vs Global

**Question:** Create global `x = 100`; inside a function create local `x = 50`. Print both and explain the difference.

````python
x = 100

def show_values():
    x = 50
    print("Inside function:", x)

show_values()
print("Outside function:", x)
````

## Q21. Using global

**Question:** Create global `count = 10` and change it to 20 inside a function using the `global` keyword.

````python
count = 10

def change_count():
    global count
    count = 20

print("Before function call:", count)
change_count()
print("After function call:", count)
````

The `global` keyword allows the function to assign a new value to the global variable.

## Q22. Integer Argument

````python
x = 10

def change_number(x):
    x = 100
    print("Inside function:", x)

print("Before function call:", x)
change_number(x)
print("After function call:", x)
````

## Q26. Calculator Function

**Question:** Create a function accepting first number, second number, and operator. Perform +, -, *, / and return the result.

````python
def calculator(first, second, operator):
    if operator == "+":
        return first + second
    elif operator == "-":
        return first - second
    elif operator == "*":
        return first * second
    elif operator == "/":
        if second != 0:
            return first / second
        return "Cannot divide by zero"
    else:
        return "Invalid operator"

print("Addition:", calculator(20, 5, "+"))
print("Subtraction:", calculator(20, 5, "-"))
print("Multiplication:", calculator(20, 5, "*"))
print("Division:", calculator(20, 5, "/"))
````

## Q27. Shopping Bill Function

**Question:** Create a function accepting item price, quantity, and discount percentage. Return the final bill amount and test with different values.

````python
def shopping_bill(price, quantity, discount_percentage):
    total = price * quantity
    discount = total * (discount_percentage / 100)
    final_amount = total - discount
    return final_amount

print("Bill 1:", shopping_bill(500, 3, 10))
print("Bill 2:", shopping_bill(1000, 5, 20))
print("Bill 3:", shopping_bill(250, 2, 5))
````

## Q29. Employee Salary System

**Question:** Create `calculate_hra()`, `calculate_da()`, and `calculate_salary()`. Use the first two functions inside the third to calculate HRA, DA, gross salary, tax, and net salary, then return the final salary details.

````python
def calculate_hra(basic_salary, hra_percentage):
    hra = basic_salary * (hra_percentage / 100)
    return hra


def calculate_da(basic_salary, da_percentage):
    da = basic_salary * (da_percentage / 100)
    return da


def calculate_salary(basic_salary, hra_percentage, da_percentage, tax_percentage):
    # Calculate HRA and DA using the first two functions
    hra = calculate_hra(basic_salary, hra_percentage)
    da = calculate_da(basic_salary, da_percentage)

    # Calculate gross salary
    gross_salary = basic_salary + hra + da

    # Calculate tax
    tax = gross_salary * (tax_percentage / 100)

    # Calculate net salary
    net_salary = gross_salary - tax

    return {
        "Basic Salary": basic_salary,
        "HRA": hra,
        "DA": da,
        "Gross Salary": gross_salary,
        "Tax": tax,
        "Net Salary": net_salary
    }


salary_details = calculate_salary(
    basic_salary=50000,
    hra_percentage=20,
    da_percentage=10,
    tax_percentage=5
)

print("Employee Salary Details:")
for key, value in salary_details.items():
    print(f"{key}: ₹{value:.2f}")
````

# Concept Checklist — Quick Revision

- **Define and call a function:** `def function_name():` followed by `function_name()`.
- **Pass arguments:** Supply values to the function parameters.
- **Required arguments:** Must be supplied.
- **Keyword arguments:** Use names such as `name="Rahul"`.
- **Default arguments:** Have a value such as `city="Bengaluru"`.
- **`*args`:** Accepts any number of positional arguments.
- **`**kwargs`:** Accepts any number of keyword arguments.
- **`return`:** Sends a result back to the caller.
- **`print()` vs `return`:** `print()` displays; `return` provides a value for further use.
- **Lambda:** A small anonymous function.
- **Local/global variables:** Local variables belong to a function; global variables are defined outside functions.
- **`global`:** Allows a function to modify a global variable.
- **Immutable vs mutable:** Integers are immutable; lists are mutable.
- **Combining functions:** One function can call another and use its returned result.
