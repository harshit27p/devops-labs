## Q10. Identity vs Equality

**Question:** For `a = [10, 20, 30]`, `b = [10, 20, 30]`, and `c = a`, predict and verify `==` and `is` results. Explain why `a == b` and `a is b` differ.

````python
a = [10, 20, 30]
b = [10, 20, 30]
c = a

print("a == b:", a == b)
print("a is b:", a is b)
print("a == c:", a == c)
print("a is c:", a is c)
````

### Expected output

```text
a == b: True
a is b: False
a == c: True
a is c: True
```

`==` compares **values/content**. `is` checks whether two variables refer to the **same object in memory**.

`a` and `b` contain the same values but are separate list objects, so `a == b` is `True` while `a is b` is `False`.

`c = a` makes `c` refer to the exact same list object as `a`, so both `a == c` and `a is c` are `True`.

## Q13. Bitwise Challenge

**Question:** Predict the output for left and right shifts of `x = 8` and explain what happens when a number is shifted.

````python
x = 8

print(x << 1)
print(x << 2)
print(x >> 1)
print(x >> 2)
````

### Expected output

```text
16
32
4
2
```

For positive integers, shifting left by one position is equivalent to multiplying by 2. Shifting right by one position is equivalent to floor-dividing by 2.

Therefore:
- `8 << 1` → 16
- `8 << 2` → 32
- `8 >> 1` → 4
- `8 >> 2` → 2

## Q16. Parentheses Challenge

**Question:** Calculate `10 + 5 * 2 ** 3`, then modify it using parentheses to produce 120, 80, and 50.

````python
# Original expression
result = 10 + 5 * 2 ** 3
print("Original:", result)

# Produces 120
result_120 = (10 + 5) * 2 ** 3
print("Produces 120:", result_120)

# Produces 80
result_80 = (10 + 5) * 2 + 50
print("Produces 80:", result_80)

# Produces 50
result_50 = (10 + 5) * 2 + 20
print("Produces 50:", result_50)
````

The original expression is:

`10 + 5 * 2 ** 3 = 50`

**Note:** The PDF asks to modify the expression using parentheses to produce 120, 80, and 50, but it does not provide a unique required expression for each target. The examples above are valid arithmetic expressions using the original numbers plus additional constants for the 80 and 50 targets.

A cleaner way to demonstrate the effect of parentheses while using only the original numbers is:
- `120` → `(10 + 5) * 2 ** 3`
- `50` → `10 + 5 * 2 ** 3`
- There is **no way to obtain 80 from only `10`, `5`, `2`, `3` using parentheses alone** without changing the operators/numbers.

## Bonus Challenge. Identity, Equality & Membership

**Question:** Predict the results for equality, identity, membership, and non-membership, then explain `==` vs `is` and `in` vs `not in`.

````python
x = 10
y = 10
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("x == y:", x == y)
print("x is y:", x is y)
print("a == b:", a == b)
print("a is b:", a is b)
print("a == c:", a == c)
print("a is c:", a is c)
print("2 in a:", 2 in a)
print("5 not in a:", 5 not in a)
````

### Expected results

```text
x == y: True
x is y: True
a == b: True
a is b: False
a == c: True
a is c: True
2 in a: True
5 not in a: True
```

**Important note:** For `x = 10` and `y = 10`, `x is y` commonly evaluates to `True` in standard CPython because small integers are typically interned/cached. However, `is` should **not** be used to test numeric value equality; use `==` for that purpose.

- `==` checks whether values are equal.
- `is` checks whether two variables refer to the same object.
- `in` checks whether a value exists in a collection.
- `not in` checks whether a value does not exist in a collection.

# Quick Revision — Operators

| Operator | Meaning |
|---|---|
| `+` | Addition / string or list concatenation |
| `-` | Subtraction |
| `*` | Multiplication / repetition |
| `/` | Division |
| `//` | Floor division |
| `%` | Remainder |
| `**` | Power |
| `>` `<` `>=` `<=` | Comparisons |
| `==` | Value equality |
| `!=` | Not equal |
| `is` | Same object identity |
| `in` | Membership |
| `not in` | Non-membership |
| `and` | True when both conditions are true |
| `or` | True when at least one condition is true |
| `not` | Reverses a Boolean result |
| `&` | Bitwise AND |
| `|` | Bitwise OR |
| `^` | Bitwise XOR |
| `~` | Bitwise NOT |
| `<<` | Left shift |
| `>>` | Right shift |
| `+=`, `-=`, `*=`, `/=` | Compound assignment |
| `**=` | Compound power assignment |

## Main precedence to remember

**Parentheses → Power → Multiplication/Division/Floor division/Modulus → Addition/Subtraction → Comparisons → `not` → `and` → `or`**
