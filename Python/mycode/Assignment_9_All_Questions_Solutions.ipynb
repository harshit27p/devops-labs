## Q1. Student Class with Constructor

**Question:** Create a class `Student` with an `__init__()` method accepting name, age, and course.
Store them as object attributes. Create two Student objects and display their details using `display_info()`.


````python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def display_info(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)


# Create two Student objects
student1 = Student("Rahul", 22, "Python")
student2 = Student("Riya", 21, "SQL")

# Display details
print("Student 1:")
student1.display_info()

print("\nStudent 2:")
student2.display_info()
````

`__init__()` is the constructor. It runs automatically when an object is created and initializes that object's attributes. `self` refers to the current object.

## Q2. Bank Account

**Question:** Create a class `BankAccount` with `account_holder` and `balance`.
Create `deposit()`, `withdraw()`, and `display_balance()` methods. Create one object, perform a deposit and withdrawal, and display the final balance.


````python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited: ₹{amount}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn: ₹{amount}")
        else:
            print("Insufficient balance.")

    def display_balance(self):
        print("Account Holder:", self.account_holder)
        print("Final Balance: ₹", self.balance)


# Create one account object
account = BankAccount("Shrishti", 5000)

print("Initial account details:")
account.display_balance()

# Deposit
account.deposit(2000)

# Withdrawal
account.withdraw(1500)

# Display final balance
print("\nFinal account details:")
account.display_balance()
````

The methods modify the object's `balance` using `self.balance`. The `withdraw()` method also checks whether enough balance is available.

## Q3. Class Variable vs Object Variable

**Question:** Create an `Employee` class with class variable `company = "ABC Technologies"`.
The constructor should accept name and salary as object variables. Create two objects and display name, salary, and company.
Then change the company name through the class and observe the result.


````python
class Employee:
    company = "ABC Technologies"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_info(self):
        print("Name:", self.name)
        print("Salary: ₹", self.salary)
        print("Company:", self.company)


# Create two Employee objects
employee1 = Employee("Rahul", 50000)
employee2 = Employee("Riya", 60000)

print("Before changing company:")
employee1.display_info()
print()
employee2.display_info()

# Change the class variable through the class
Employee.company = "XYZ Technologies"

print("\nAfter changing company through the class:")
employee1.display_info()
print()
employee2.display_info()
````

`company` is a **class variable**, shared by objects unless an object has its own attribute with the same name. `name` and `salary` are **object variables**, so each employee has separate values.

## Q4. Using self with Object Methods

**Question:** Create a `Rectangle` class with object attributes `length` and `breadth`.
Create `area()` and `perimeter()` methods that return the respective values using `self`.
Create two Rectangle objects with different dimensions and display their areas and perimeters.


````python
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length + self.breadth)


# Create two Rectangle objects
rectangle1 = Rectangle(10, 5)
rectangle2 = Rectangle(8, 4)

print("Rectangle 1:")
print("Length:", rectangle1.length)
print("Breadth:", rectangle1.breadth)
print("Area:", rectangle1.area())
print("Perimeter:", rectangle1.perimeter())

print("\nRectangle 2:")
print("Length:", rectangle2.length)
print("Breadth:", rectangle2.breadth)
print("Area:", rectangle2.area())
print("Perimeter:", rectangle2.perimeter())
````

`self.length` and `self.breadth` refer to the dimensions stored in the current Rectangle object. The methods use those values to calculate area and perimeter.

## Q5. Product Inventory

**Question:** Create a `Product` class with class variable `category = "Electronics"` and object variables `name`, `price`, and `quantity`.
Create `display_product()` and `total_value()` methods. Create two product objects and display their details and total inventory value.


````python
class Product:
    category = "Electronics"

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def display_product(self):
        print("Product Name:", self.name)
        print("Category:", self.category)
        print("Price: ₹", self.price)
        print("Quantity:", self.quantity)

    def total_value(self):
        return self.price * self.quantity


# Create two product objects
product1 = Product("Laptop", 50000, 2)
product2 = Product("Headphones", 2000, 5)

print("Product 1:")
product1.display_product()
print("Total Value: ₹", product1.total_value())

print("\nProduct 2:")
product2.display_product()
print("Total Value: ₹", product2.total_value())

# Total value of the complete inventory
total_inventory_value = product1.total_value() + product2.total_value()

print("\nTotal Inventory Value: ₹", total_inventory_value)
````

`category` is a class variable shared by all Product objects. `name`, `price`, and `quantity` are object variables. `total_value()` calculates `price × quantity` for one product.

## Q6. Challenge — Student Result

**Question:** Create a `Student` class with object variables `name` and `marks`, where `marks` is a list.
Create methods to calculate total and average marks. Create two objects and display their names, total marks, and average marks.
Also create class variable `school = "ABC School"` and display it for both objects.


````python
class Student:
    school = "ABC School"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def total_marks(self):
        return sum(self.marks)

    def average_marks(self):
        return self.total_marks() / len(self.marks)

    def display_result(self):
        print("Name:", self.name)
        print("School:", self.school)
        print("Marks:", self.marks)
        print("Total Marks:", self.total_marks())
        print("Average Marks:", self.average_marks())


# Create two Student objects
student1 = Student("Rahul", [85, 90, 80])
student2 = Student("Riya", [92, 88, 95])

# Display results
print("Student 1:")
student1.display_result()

print("\nStudent 2:")
student2.display_result()
````

`school` is a class variable, so both objects can access the same value. `name` and `marks` are object variables, so each student can have different values.

# Concept Checklist — Quick Revision

### 1. Can I create a class using the `class` statement?
Yes.

```python
class Student:
    pass
```

### 2. Can I create objects from a class?
Yes.

```python
student1 = Student("Rahul", 22, "Python")
```

### 3. What is the purpose of `self`?
`self` refers to the current object. It is used to access that object's attributes and methods.

### 4. Can I define and use object methods?
Yes.

```python
def display_info(self):
    print(self.name)
```

### 5. Can I use `__init__()` to initialize object attributes?
Yes. `__init__()` runs automatically when an object is created.

### 6. Class variables vs object variables
- **Class variable:** Shared by the class/objects, e.g. `company = "ABC Technologies"`.
- **Object variable:** Belongs to an individual object, e.g. `self.name`.

### 7. Can I access class and object variables using an object?
Yes.

```python
employee1.company
employee1.name
```

### 8. Can I create multiple objects with different attribute values?
Yes. Each object can store its own object-variable values.
