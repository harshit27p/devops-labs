## Q1. Student Class

**Question:** Create a class `Student` with the attributes `name`, `age`, and `course`.
Create one student object and print all its attributes.

````python
class Student:
    pass


# Create a student object
student1 = Student()

# Assign attributes
student1.name = "Rahul"
student1.age = 22
student1.course = "Python"

# Print all attributes
print("Name:", student1.name)
print("Age:", student1.age)
print("Course:", student1.course)
````

**Concept:** An attribute is data stored inside an object. Here, `name`, `age`, and `course` are attributes of `student1`.

## Q2. Car Class

**Question:** Create a class `Car` with `brand`, `model`, and `color`.
Create two car objects with different values and print their details.

````python
class Car:
    pass


# First car object
car1 = Car()
car1.brand = "Toyota"
car1.model = "Camry"
car1.color = "White"

# Second car object
car2 = Car()
car2.brand = "Honda"
car2.model = "City"
car2.color = "Black"

# Print details
print("Car 1:")
print("Brand:", car1.brand)
print("Model:", car1.model)
print("Color:", car1.color)

print("\nCar 2:")
print("Brand:", car2.brand)
print("Model:", car2.model)
print("Color:", car2.color)
````

Both objects are created from the same `Car` class, but each object has its own attribute values.

## Q3. Method in a Class

**Question:** Create a class `Person` with the attribute `name` and a method `introduce()`.
The method should print `Hello, my name is Rahul.` Create an object and call the method.

````python
class Person:
    def introduce(self):
        print(f"Hello, my name is {self.name}.")


# Create an object
person1 = Person()

# Assign the name attribute
person1.name = "Rahul"

# Call the method
person1.introduce()
````

A **method** is a function defined inside a class. `self` refers to the current object, so `self.name` accesses that object's `name` attribute.

## Q4. Student Method

**Question:** Create a class `Student` with attributes `name` and `marks`, and a method `display_result()`.
The method should print:
`Student: Rahul`
`Marks: 85`
Create an object and call the method.

````python
class Student:
    def display_result(self):
        print("Student:", self.name)
        print("Marks:", self.marks)


# Create an object
student1 = Student()

# Assign attributes
student1.name = "Rahul"
student1.marks = 85

# Call the method
student1.display_result()
````

`display_result()` is a method because it is defined inside the `Student` class. It uses `self` to access the object's attributes.

## Q5. Simple Bank Account

**Question:** Create a class `BankAccount` with attributes `account_holder` and `balance`, and a method `display_balance()`.
Create an object with:
- Account Holder: Shrishti
- Balance: 5000

Call the method to display the account details.

````python
class BankAccount:
    def display_balance(self):
        print("Account Holder:", self.account_holder)
        print("Balance:", self.balance)


# Create a bank account object
account1 = BankAccount()

# Assign attributes
account1.account_holder = "Shrishti"
account1.balance = 5000

# Call the method
account1.display_balance()
````

The `BankAccount` object contains two attributes: `account_holder` and `balance`. The `display_balance()` method accesses and displays those attributes.

# Concept Checklist — Quick Revision

### 1. Can I create a class?
Use the `class` keyword:

```python
class Student:
    pass
```

### 2. Can I create an object from a class?
Create an object by calling the class:

```python
student1 = Student()
```

### 3. What is an attribute?
An attribute is data associated with an object:

```python
student1.name = "Rahul"
```

### 4. Can I access an object's attributes?
Use the dot operator:

```python
print(student1.name)
```

### 5. Can I define a method inside a class?
Yes:

```python
class Person:
    def introduce(self):
        print("Hello")
```

### 6. Can I call a method using an object?
Yes:

```python
person1.introduce()
```

### 7. Can I create multiple objects from the same class?
Yes. For example:

```python
car1 = Car()
car2 = Car()
```

Each object can have different attribute values.
