## Q17. Student Marks

Create marks for five subjects and calculate total marks, number of subjects, average, highest, and lowest marks.

```python
marks = [85, 72, 90, 65, 88]

total_marks = sum(marks)
number_of_subjects = len(marks)
average_marks = total_marks / number_of_subjects
highest_marks = max(marks)
lowest_marks = min(marks)

print("Marks:", marks)
print("Total marks:", total_marks)
print("Number of subjects:", number_of_subjects)
print("Average marks:", average_marks)
print("Highest marks:", highest_marks)
print("Lowest marks:", lowest_marks)
```

# Concept Checklist — Quick Revision

- **Variable:** A name used to store a value.
- **`print()`**: Displays output on the screen.
- **`input()`**: Takes input from the user and normally returns a string.
- **Convert input:** Use `int()` for integers and `float()` for decimal numbers.
- **`int` vs `float`:** `int` stores whole numbers; `float` stores numbers with decimal values.
- **Arithmetic operators:** `+`, `-`, `*`, `/`, `%`, `//`, `**`.
- **List:** An ordered, mutable collection of values.
- **List index:** Position used to access an element; indexing starts at `0`.
- **Index `0`:** Refers to the first element.
- **Modify a list element:** Assign a new value using its index, e.g. `colors[1] = "Black"`.
- **`append()` vs `insert()`:** `append()` adds at the end; `insert(index, value)` adds at a specified position.
- **`remove()`:** Removes the first matching value from a list.
- **`len()` / `sum()` / `min()` / `max()`:** Return the number of elements / total / smallest value / largest value.
