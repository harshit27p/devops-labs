## Q6. Trigonometric Functions

**Question:** Take an angle in degrees, convert it to radians, and calculate sine, cosine, and tangent. Test with 0, 30, 45, 60, and 90 degrees.

```python
import math

test_angles = [0, 30, 45, 60, 90]

for angle in test_angles:
    radians = math.radians(angle)
    print(f"Angle: {angle}°")
    print("Radians:", radians)
    print("Sine:", math.sin(radians))
    print("Cosine:", math.cos(radians))

    # tan(90°) is mathematically undefined; Python's floating-point
    # calculation gives a very large value instead.
    if angle == 90:
        print("Tangent: undefined (90°)")
    else:
        print("Tangent:", math.tan(radians))
    print("-" * 30)

# User-input version:
# angle = float(input("Enter an angle in degrees: "))
# radians = math.radians(angle)
# print("Sine:", math.sin(radians))
# print("Cosine:", math.cos(radians))
# print("Tangent:", math.tan(radians))
```

## Q36. Set Operations

**Question:** Find students studying both Python and SQL, either Python or SQL, Python but not SQL, and SQL but not Python.

```python
python_students = {"Aman", "Riya", "Karan", "Neha"}
sql_students = {"Riya", "Neha", "Arjun", "Priya"}

both = python_students & sql_students
either = python_students | sql_students
python_only = python_students - sql_students
sql_only = sql_students - python_students

print("Both Python and SQL:", both)
print("Either Python or SQL:", either)
print("Python but not SQL:", python_only)
print("SQL but not Python:", sql_only)
```

## Q38. List vs Tuple

**Question:** Compare a list and tuple by accessing the second element, finding length, trying to modify the second element, and explaining the difference.

```python
my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

print("List second element:", my_list[1])
print("Tuple second element:", my_tuple[1])

print("List length:", len(my_list))
print("Tuple length:", len(my_tuple))

my_list[1] = 200
print("Modified list:", my_list)

try:
    my_tuple[1] = 200
except TypeError as error:
    print("Tuple modification error:", error)
```

**Key difference:** Lists are mutable, so their elements can be changed. Tuples are immutable, so their elements cannot be changed after creation.

# Concept Check — Answers

These questions are included on pages 17–18 of the assignment PDF.

### 1. What is the difference between int, float, and complex?

`int` stores whole numbers, `float` stores decimal numbers, and `complex` stores numbers with real and imaginary parts, such as `3 + 4j`.

### 2. How do you convert one number type into another?

Use conversion functions such as `int()`, `float()`, and `complex()`, when the value is valid for that conversion.

### 3. What is the difference between /, //, %, and **?

`/` performs normal division, `//` performs floor division, `%` gives the remainder, and `**` performs exponentiation (power).

### 4. What does the math module provide?

It provides mathematical functions and constants such as `sqrt()`, `ceil()`, `floor()`, `sin()`, `pi`, and `e`.

### 5. What is the purpose of the random module?

It provides functions for generating pseudo-random values, such as random integers, random floating-point numbers, and random selections.

### 6. What does it mean that strings are immutable?

A string cannot be changed in place after it is created. Operations that appear to change a string create a new string.

### 7. How does string indexing work?

Indexing accesses individual characters. Indexing starts at `0`; negative indexes count from the end. For example, `text[0]` is the first character and `text[-1]` is the last.

### 8. How does string slicing work?

Slicing extracts a sequence using `start:stop:step`. The stop index is normally excluded. For example, `text[:5]` gives the first five characters.

### 9. What are escape characters?

Escape sequences represent special characters inside strings. Examples include `\n` for a new line, `\t` for a tab, and `\"` for a double quote.

### 10. What is the difference between + and * when used with strings?

`+` concatenates strings, while `*` repeats a string a specified number of times.

### 11. What is an f-string?

An f-string is a formatted string literal prefixed with `f`. It allows expressions or variables to be inserted using `{}`.

### 12. What is the difference between a list and a tuple?

Both are ordered collections, but lists are mutable and tuples are immutable.

### 13. What is the difference between remove(), pop(), and del?

`remove(value)` removes a matching value from a list; `pop(index)` removes and returns an element (last by default); `del` deletes an element, slice, or variable.

### 14. Can tuples be modified?

No. Tuples are immutable, so individual tuple elements cannot be changed after creation.

### 15. What is a dictionary key?

A dictionary key is the unique identifier used to access a value in a dictionary, for example `student['name']`.

### 16. Why can't a list normally be used as a dictionary key?

A list is mutable and unhashable, so it cannot normally be used as a dictionary key.

### 17. What is the difference between get() and direct dictionary access?

`dictionary[key]` raises a `KeyError` when the key is absent. `dictionary.get(key)` returns `None` by default (or a supplied default value) when the key is absent.

### 18. Why are duplicate values removed from a set?

A set stores unique elements only. When duplicate values are supplied, only one copy of each value remains.

### 19. What is the difference between union and intersection?

Union contains elements from both sets. Intersection contains only elements common to both sets.

### 20. What is the difference between a set and a list?

A list is ordered, supports indexing, and can contain duplicates. A set stores unique elements and does not support index-based access.

### 21. Which of these data types are mutable and immutable?

Commonly used here: **mutable** — list, dictionary, set. **Immutable** — int, float, complex, string, tuple.
