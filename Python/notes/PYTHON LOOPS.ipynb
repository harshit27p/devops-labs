# Python Loops

Loops are used to **repeat a block of code multiple times**.

Instead of writing the same statement again and again, a loop allows Python to execute it repeatedly based on a condition or over the elements of an iterable.

Python mainly provides:

- `while` loop
- `for` loop

Loops can also be combined to create **nested loops**.

Python provides loop-control statements to change the normal flow of a loop:

- `break`
- `continue`
- `pass`


## Various Types of Loops

Python's main loop structures are:

| Loop | Purpose |
|---|---|
| `while` loop | Repeats while a condition is `True` |
| `for` loop | Iterates over the elements of an iterable |
| Nested loop | A loop placed inside another loop |

### Loop Control Statements

| Statement | Purpose |
|---|---|
| `break` | Immediately terminates the loop |
| `continue` | Skips the current iteration and moves to the next one |
| `pass` | Does nothing; acts as a placeholder |


## `while` Loop

A `while` loop repeatedly executes a block of code **as long as its condition remains `True`**.

### Syntax

```python
while condition:
    statement(s)
```

The condition is checked **before every iteration**.

If the condition becomes `False`, the loop stops.


````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

### How a `while` Loop Works

For the example above:

1. `count` starts at `1`.
2. Python checks `count <= 5`.
3. If the condition is `True`, the loop body executes.
4. `count` is increased by `1`.
5. The condition is checked again.
6. When `count` becomes `6`, the condition is `False`, so the loop stops.


### Important: Updating the Loop Variable

A `while` loop must usually have something inside it that eventually makes the condition `False.

Otherwise, the loop may continue forever. This is called an **infinite loop**.


````python
count = 1

while count <= 3:
    print(count)
    count += 1
````

### Example: Sum of Numbers Using `while`

A `while` loop can be used when you need to repeat an operation until a condition is satisfied.


````python
number = 1
total = 0

while number <= 5:
    total += number
    number += 1

print("Sum:", total)
````

## `for` Loop

A `for` loop is used to iterate over the elements of an **iterable**.

Common iterables include:

- Lists
- Tuples
- Strings
- Sets
- Dictionaries
- `range()`

### Syntax

```python
for variable in iterable:
    statement(s)
```

The loop variable receives one element at a time.


````python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
````

### `for` Loop with a String

A string is an iterable, so a `for` loop can process it one character at a time.


````python
word = "Python"

for character in word:
    print(character)
````

### `for` Loop with `range()`

`range()` is commonly used when you want to repeat a block a specific number of times.

```python
range(start, stop, step)
```

The `stop` value is excluded.


````python
for number in range(1, 6):
    print(number)
````

### `range()` with a Step

The third argument specifies the step size.


````python
for number in range(0, 11, 2):
    print(number)
````

### `for` Loop with a Dictionary

When you directly iterate over a dictionary, the loop variable receives its **keys**.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "course": "Python"
}

for key in student:
    print(key, ":", student[key])
````

### Iterating Through Dictionary Items

The `items()` method can be used when both the key and value are required.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

for key, value in student.items():
    print(key, ":", value)
````

## `while` Loop vs `for` Loop

| Feature | `while` | `for` |
|---|---|---|
| Main idea | Repeat while a condition is true | Iterate over an iterable |
| Best when | Number of iterations is condition-dependent | Iterating through a collection or known range |
| Condition | Explicitly written | Controlled by iterable |
| Loop variable update | Usually done manually | Handled by iteration |
| Common use | Input validation, condition-based repetition | Lists, strings, ranges, dictionaries |

### Simple rule

> Use `for` when you are iterating over something.

> Use `while` when you want to continue until a condition changes.


## Nested Loop

A **nested loop** is a loop placed inside another loop.

The inner loop executes completely for each iteration of the outer loop.

### Syntax

```python
for outer_variable in outer_iterable:
    for inner_variable in inner_iterable:
        statement(s)
```


````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
````

### Understanding the Execution

For:

```python
for i in range(1, 4):
    for j in range(1, 4):
```

the inner loop runs three times for every value of `i`.

So the total number of inner-loop executions is:

```text
3 × 3 = 9
```


### Example: Multiplication Table

Nested loops can be used to generate tables or matrix-like output.


````python
for i in range(1, 4):
    for j in range(1, 6):
        print(i * j, end=" ")
    print()
````

### Nested `while` Loops

Nested loops are not limited to `for` loops. A `while` loop can also be placed inside another `while` loop.


````python
i = 1

while i <= 3:
    j = 1

    while j <= 2:
        print("i =", i, "j =", j)
        j += 1

    i += 1
````

## Loop Control Statements

Loop control statements change the normal execution of a loop.

Python provides:

- `break`
- `continue`
- `pass`


## `break` Statement

The `break` statement **immediately terminates the nearest enclosing loop**.

Execution continues with the statement after the loop.


````python
for number in range(1, 6):
    if number == 4:
        break

    print(number)

print("Loop ended")
````

### How `break` Works

In the example:

- `1`, `2`, and `3` are printed.
- When `number == 4`, `break` executes.
- The loop stops immediately.
- `5` is never processed.


### `break` in a `while` Loop

`break` can also terminate a `while` loop before its condition becomes false.


````python
count = 1

while count <= 10:
    if count == 5:
        break

    print(count)
    count += 1
````

## `continue` Statement

The `continue` statement **skips the remaining statements in the current iteration** and moves to the next iteration of the loop.

Unlike `break`, `continue` does **not** terminate the loop.


````python
for number in range(1, 6):
    if number == 3:
        continue

    print(number)
````

### `break` vs `continue`

| Statement | What happens? |
|---|---|
| `break` | Terminates the entire loop |
| `continue` | Skips only the current iteration |

For example, with numbers `1` to `5`:

```text
break at 3    → 1 2
continue at 3 → 1 2 4 5
```


### `continue` in a `while` Loop

Be careful when using `continue` in a `while` loop.

The loop-control variable should be updated before `continue`, otherwise the condition may never change and an infinite loop can occur.


````python
count = 0

while count < 5:
    count += 1

    if count == 3:
        continue

    print(count)
````

## `pass` Statement

The `pass` statement **does nothing**.

It is used as a placeholder when Python requires a statement syntactically, but you do not want any action to occur yet.

Unlike `break` and `continue`, `pass` does not change the loop's control flow.


````python
for number in range(1, 4):
    if number == 2:
        pass
    print(number)
````

### `pass` vs `continue`

This is an important distinction:

- `pass` → does nothing; execution continues normally.
- `continue` → skips the remaining part of the current iteration.


````python
print("Using pass:")

for number in range(1, 4):
    if number == 2:
        pass
    print(number)

print("\nUsing continue:")

for number in range(1, 4):
    if number == 2:
        continue
    print(number)
````

## Iterator

An **iterator** is an object that produces values one at a time.

An iterator follows Python's iterator protocol:

- `__iter__()` returns an iterator.
- `__next__()` returns the next value.
- When there are no more values, `__next__()` raises `StopIteration`.

Many Python objects, such as lists, tuples, strings, and sets, are **iterable** and can be used to create iterators.


````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

### `StopIteration`

After an iterator has produced all its values, calling `next()` again raises `StopIteration`.


````python
numbers = [10, 20]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))

try:
    print(next(iterator))
except StopIteration:
    print("No more values")
````

### Iterable vs Iterator

These terms are related but not identical.

**Iterable**
- An object that can be iterated over.
- Can generally be passed to `iter()`.
- Examples: list, tuple, string, set, dictionary.

**Iterator**
- An object that produces the next value using `next()`.
- Keeps track of its current position/state.


````python
numbers = [10, 20, 30]

# List is iterable
print(iter(numbers))

# The iterator is obtained from the iterable
iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
````

## Generator

A **generator** is a special type of iterator that generates values **one at a time**, usually using the `yield` keyword.

Generators are useful when you do not want to create and store all values in memory at once.

### Generator Function

A function containing `yield` becomes a generator function.


````python
def count_up_to(n):
    number = 1

    while number <= n:
        yield number
        number += 1

numbers = count_up_to(3)

print(next(numbers))
print(next(numbers))
print(next(numbers))
````

### Using a Generator with a `for` Loop

Generators can be directly used in a `for` loop.


````python
def count_up_to(n):
    number = 1

    while number <= n:
        yield number
        number += 1

for number in count_up_to(5):
    print(number)
````

### `return` vs `yield`

`return` ends a function and sends back a result.

`yield` produces a value and **pauses the generator's execution**. When the generator is requested for another value, it resumes from where it stopped.

This allows generators to produce values lazily.


````python
def simple_generator():
    print("Starting")
    yield 1

    print("Resuming")
    yield 2

generator = simple_generator()

print(next(generator))
print(next(generator))
````

### Generator Expression

A generator can also be created using a generator expression.

It looks similar to a list comprehension, but uses parentheses instead of square brackets.


````python
numbers = (x * x for x in range(1, 6))

for number in numbers:
    print(number)
````

### Why Use Generators?

Generators are useful when:

- The data is large.
- Values can be produced one at a time.
- You want to avoid storing the entire result in memory.
- You are working with streams or sequences of values.

A generator provides values **lazily**, meaning values are produced when they are requested.


## Iterator vs Generator

| Feature | Iterator | Generator |
|---|---|---|
| Meaning | Object that produces values one at a time | Special kind of iterator |
| Created using | `iter()` or custom iterator class | `yield` or generator expression |
| Uses `next()` | Yes | Yes |
| Maintains state | Yes | Yes |
| Lazy | Yes | Yes |
| Code required | Can require more implementation | Usually simpler |
| Typical use | Custom iteration behavior | Efficient/lazy value generation |

### Key relationship

> **Every generator is an iterator, but not every iterator is a generator.**
