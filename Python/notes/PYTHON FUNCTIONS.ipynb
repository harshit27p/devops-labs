# Python Functions

A **function** is a reusable block of code designed to perform a specific task.

Functions help us:
- Avoid repeating code
- Organize programs into smaller logical parts
- Reuse the same logic with different inputs
- Make programs easier to read and maintain

Python provides built-in functions such as `print()`, `len()`, and `type()`, and we can also create our own functions.


## Defining a Function

A function is defined using the `def` keyword.

### Syntax

```python
def function_name(parameters):
    statement(s)
```

- `def` → keyword used to define a function
- `function_name` → name of the function
- `parameters` → optional inputs
- `:` → starts the function body
- Indented statements → function body


````python
def greet():
    print("Hello, welcome to Python!")

print("Function defined successfully")
````

## Calling a Function

Defining a function does not execute it.

The function runs when it is **called**.

### Syntax

```python
function_name()
```


````python
def greet():
    print("Hello!")

greet()
greet()
````

### Function with a Parameter

A parameter allows a function to receive input.

The value supplied when calling the function is called an **argument**.


````python
def greet(name):
    print("Hello", name)

greet("Shrishti")
greet("Rahul")
````

## Function Arguments

Arguments are values passed to a function when it is called.

Python supports:
- Required / positional arguments
- Keyword arguments
- Default arguments
- Variable-length arguments


## Required Arguments

A required argument must be supplied when the function is called.

If a required argument is missing, Python raises a `TypeError`.


````python
def add(a, b):
    print(a + b)

add(10, 20)

# add(10)  # TypeError: missing required argument
````

## Keyword Arguments

Keyword arguments pass values using the **parameter name**.

The order of keyword arguments does not have to match the parameter order.


````python
def student_info(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student_info(
    course="Python",
    name="Shrishti",
    age=23
)
````

### Positional vs Keyword Arguments

**Positional:**

```python
student_info("Shrishti", 23, "Python")
```

Values are matched according to position.

**Keyword:**

```python
student_info(name="Shrishti", age=23, course="Python")
```

Values are matched using parameter names.


## Default Arguments

A **default argument** is a parameter that already has a default value.

If the caller does not provide that argument, Python uses the default.

### Syntax

```python
def function_name(parameter=default_value):
    ...
```


````python
def greet(name, message="Good morning"):
    print(message, name)

greet("Shrishti")
greet("Shrishti", "Good evening")
````

### Important Rule for Default Arguments

Parameters with default values must come **after** parameters without default values.

Correct:

```python
def example(a, b=10):
    ...
```

Incorrect:

```python
# def example(a=10, b):
#     ...
```


## Variable-length Arguments

Sometimes we do not know in advance how many arguments a function will receive.

Python provides:
- `*args` → variable number of positional arguments
- `**kwargs` → variable number of keyword arguments


### `*args`

`*args` collects extra positional arguments into a **tuple**.


````python
def add_numbers(*args):
    total = 0

    for number in args:
        total += number

    return total

print(add_numbers(10, 20))
print(add_numbers(10, 20, 30, 40))
````

### `**kwargs`

`**kwargs` collects extra keyword arguments into a **dictionary**.


````python
def display_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

display_info(name="Shrishti", age=23, city="Bengaluru")
````

### `*args` vs `**kwargs`

| Feature | `*args` | `**kwargs` |
|---|---|---|
| Accepts | Extra positional arguments | Extra keyword arguments |
| Stored as | Tuple | Dictionary |
| Example | `func(10, 20, 30)` | `func(a=10, b=20)` |


## The Anonymous Functions

An **anonymous function** is a function without a regular function name.

Python creates anonymous functions using the `lambda` keyword.

### Syntax

```python
lambda arguments: expression
```

A lambda contains one expression, and that expression is automatically returned.


````python
square = lambda x: x * x

print(square(5))
print(square(8))
````

### Lambda with Multiple Arguments

A lambda can accept multiple arguments.


````python
add = lambda a, b: a + b

print(add(10, 20))
````

### Lambda with `sorted()`

Lambda functions are useful when a small function is needed temporarily, such as providing a sorting key.


````python
students = [
    ("Aman", 82),
    ("Shrishti", 91),
    ("Riya", 76)
]

students_sorted = sorted(students, key=lambda student: student[1])

print(students_sorted)
````

## The `return` Statement

The `return` statement sends a value back to the code that called the function.

It also **ends the function immediately**.


````python
def add(a, b):
    result = a + b
    return result

answer = add(10, 20)

print(answer)
````

### Returning Multiple Values

Python can return multiple values from a function. They are returned together as a tuple.


````python
def calculate(a, b):
    total = a + b
    difference = a - b

    return total, difference

result = calculate(10, 3)

print(result)
````

### `print()` vs `return`

- `print()` displays a value on the screen.
- `return` sends a value back to the caller so it can be stored or used later.


````python
def using_print():
    print("Hello")

def using_return():
    return "Hello"

result = using_print()
print("Returned value:", result)

result = using_return()
print("Returned value:", result)
````

## Scope of Variables

**Scope** refers to the part of a program where a variable can be accessed.

The two important scopes here are:
- Local scope
- Global scope


## Local Variables

A variable created **inside a function** is generally a local variable.

It can normally be accessed only inside that function.


````python
def calculate():
    x = 10
    print("Inside function:", x)

calculate()

# x is not available here because it is local to calculate()
````

## Global Variables

A variable created **outside a function** is a global variable.

It can be accessed from inside a function when the function does not create a local variable with the same name.


````python
x = 100

def display():
    print("Inside function:", x)

display()
print("Outside function:", x)
````

## Global vs Local Variables

If a local variable has the same name as a global variable, the local variable is used inside the function.


````python
x = 100

def display():
    x = 50
    print("Inside function:", x)

display()

print("Outside function:", x)
````

### Using the `global` Keyword

If a function needs to **rebind a global variable**, use the `global` keyword.

Without `global`, assignment inside the function creates a local variable.


````python
count = 0

def increment():
    global count
    count += 1

increment()
increment()

print(count)
````

### Avoid Unnecessary Global Variables

Although `global` is available, excessive global state can make programs harder to understand and maintain.

For most programs, it is usually better to pass data through parameters and return results.


## Pass by Reference vs Value

Python is best described as using **object-reference / object-sharing semantics** for function arguments.

The function parameter receives a reference to the same object.

What happens next depends on whether the object is mutable or immutable.

- Immutable examples: `int`, `float`, `str`, `tuple`
- Mutable examples: `list`, `dict`, `set`


### Immutable Object Example

Integers are immutable.

If the parameter is reassigned inside the function, the original variable outside the function is not changed.


````python
def change_number(x):
    x = x + 10
    print("Inside function:", x)

number = 5

change_number(number)

print("Outside function:", number)
````

### Mutable Object Example

Lists are mutable.

A function can modify the contents of the same list object, and that change can be visible outside the function.


````python
def add_item(items):
    items.append("Python")

languages = ["SQL", "Java"]

add_item(languages)

print(languages)
````

### Reassigning a Mutable Object

Mutating an object and rebinding the parameter are different.

Reassigning the parameter to a new list does not replace the caller's original list.


````python
def replace_list(items):
    items = ["Python", "SQL"]
    print("Inside function:", items)

languages = ["Java", "C++"]

replace_list(languages)

print("Outside function:", languages)
````

### Key Idea

Do not think of Python simply as "pass by value" or "pass by reference."

A better mental model is:

> **Python passes object references by assignment.**

The function parameter and the caller's variable can refer to the same object. Mutating that object can therefore be visible outside the function, while rebinding the parameter does not rebind the caller's variable.


## Function Arguments — Quick Comparison

| Type | Example | Main Idea |
|---|---|---|
| Required / positional | `func(10, 20)` | Arguments supplied according to position |
| Keyword | `func(a=10, b=20)` | Arguments supplied by parameter name |
| Default | `def func(a=10)` | Uses a default when no value is supplied |
| `*args` | `func(10, 20, 30)` | Collects extra positional arguments into a tuple |
| `**kwargs` | `func(a=10, b=20)` | Collects extra keyword arguments into a dictionary |
