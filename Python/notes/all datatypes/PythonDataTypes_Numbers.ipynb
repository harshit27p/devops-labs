# Python Data Types – Numbers

Python provides three main built-in numeric data types:

- `int` → Integer numbers
- `float` → Decimal / floating-point numbers
- `complex` → Complex numbers

Python also provides several modules and functions for performing mathematical, random, and trigonometric operations.


## `int`

The `int` data type is used to represent **whole numbers**, both positive and negative.

Examples:

```python
10
-25
0
1000
```

Integers do not contain a decimal point.


````python
age = 23
temperature = -5
count = 0

print(age)
print(temperature)
print(count)

print(type(age))
````

### Integer Arithmetic

Integers can be used with normal arithmetic operators such as:

- `+` → Addition
- `-` → Subtraction
- `*` → Multiplication
- `/` → Division
- `//` → Floor division
- `%` → Modulus
- `**` → Exponentiation


````python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

## `float`

The `float` data type represents **numbers containing a decimal point**.

Examples:

```python
3.14
-2.5
0.0
99.99
```

Division using `/` generally produces a float, even when both operands are integers.


````python
price = 99.99
temperature = -2.5
value = 10.0

print(price)
print(temperature)
print(value)

print(type(price))
````

### Floating-Point Arithmetic

Floating-point numbers can be used with the same arithmetic operators as integers.

Because floating-point numbers are represented approximately in computer memory, some calculations may produce results with very small precision differences.


````python
a = 10.5
b = 2.5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
````

## `complex`

The `complex` data type represents **complex numbers**.

A complex number has two parts:

```text
real part + imaginary part
```

Python uses `j` to represent the imaginary part.

### Syntax

```python
a + bj
```

Example:

```python
z = 3 + 4j
```

Here:

- `3` → real part
- `4` → imaginary part


````python
z = 3 + 4j

print(z)
print(type(z))

print(z.real)
print(z.imag)
````

### Creating Complex Numbers Using `complex()`

The `complex()` function can also be used to create a complex number.

```python
complex(real, imaginary)
```


````python
z1 = complex(3, 4)
z2 = complex(5, -2)

print(z1)
print(z2)
````

### Complex Number Operations

Complex numbers support arithmetic operations such as addition, subtraction, multiplication, and division.



````python
z1 = 2 + 3j
z2 = 1 + 2j

print(z1 + z2)
print(z1 - z2)
print(z1 * z2)
print(z1 / z2)
````

## Number Type Conversions

Python provides functions for converting between numeric types.

Common conversion functions are:

- `int()` → converts a value to an integer
- `float()` → converts a value to a floating-point number
- `complex()` → creates a complex number


### `int()`

`int()` converts a compatible value into an integer.

When converting a float to an integer, the decimal portion is **discarded**; it is not rounded.


````python
print(int(10.8))
print(int(-5.9))

print(int("25"))
````

### `float()`

`float()` converts a compatible value into a floating-point number.


````python
print(float(10))
print(float("25.5"))

number = 20
print(float(number))
````

### `complex()`

`complex()` can create a complex number from a real value or from real and imaginary components.


````python
print(complex(5))
print(complex(3, 4))
````

### Important Conversion Point

Some conversions are not directly possible.

For example, converting a string containing a decimal value directly to `int()` raises an error:

```python
int("10.5")
```

Instead, convert it to `float` first and then to `int` if that behavior is intended.


````python
value = "10.5"

number = int(float(value))

print(number)
````

## Mathematical Functions

Python's `math` module provides many mathematical functions.

First import the module:

```python
import math
```

Some commonly used functions are:

- `ceil()`
- `floor()`
- `sqrt()`
- `pow()`
- `fabs()`
- `factorial()`


### `ceil()`

`math.ceil(x)` returns the **smallest integer greater than or equal to `x`**.


````python
import math

print(math.ceil(4.2))
print(math.ceil(4.9))
print(math.ceil(-4.2))
````

### `floor()`

`math.floor(x)` returns the **largest integer less than or equal to `x`**.


````python
import math

print(math.floor(4.2))
print(math.floor(4.9))
print(math.floor(-4.2))
````

### `sqrt()`

`math.sqrt(x)` returns the square root of `x`.

For real-number calculations, the input should be non-negative.


````python
import math

print(math.sqrt(25))
print(math.sqrt(2))
````

### `pow()`

`math.pow(x, y)` returns `x` raised to the power `y` as a floating-point result.

Python also has the `**` operator for exponentiation.


````python
import math

print(math.pow(2, 3))
print(2 ** 3)
````

### `fabs()`

`math.fabs(x)` returns the absolute value of `x` as a float.


````python
import math

print(math.fabs(-10))
print(math.fabs(5.5))
````

### `factorial()`

`math.factorial(n)` returns the factorial of a non-negative integer.

For example:

```text
5! = 5 × 4 × 3 × 2 × 1 = 120
```


````python
import math

print(math.factorial(5))
print(math.factorial(6))
````

### Other Useful Mathematical Functions

The `math` module also provides functions such as:

- `gcd()` → greatest common divisor
- `lcm()` → least common multiple
- `isqrt()` → integer square root
- `log()` → logarithm
- `log10()` → base-10 logarithm
- `exp()` → exponential function


````python
import math

print(math.gcd(12, 18))
print(math.lcm(4, 6))
print(math.isqrt(25))
print(math.log10(100))
print(math.exp(1))
````

## Random Number Functions

Python provides the `random` module for generating pseudo-random values.

First import it:

```python
import random
```

Common functions include:

- `random()`
- `randint()`
- `randrange()`
- `uniform()`
- `choice()`
- `shuffle()`


### `random()`

`random.random()` returns a random floating-point number in the range:

```text
0.0 <= value < 1.0
```


````python
import random

print(random.random())
print(random.random())
````

### `randint()`

`random.randint(a, b)` returns a random integer **including both `a` and `b`**.


````python
import random

print(random.randint(1, 10))
print(random.randint(100, 200))
````

### `randrange()`

`random.randrange()` returns a randomly selected value from a range.

The stop value is excluded, similar to `range()`.


````python
import random

print(random.randrange(1, 10))
print(random.randrange(0, 20, 2))
````

### `uniform()`

`random.uniform(a, b)` returns a random floating-point number between `a` and `b`.


````python
import random

print(random.uniform(1.5, 5.5))
````

### `choice()`

`random.choice()` selects one random element from a sequence.


````python
import random

fruits = ["apple", "banana", "mango", "orange"]

print(random.choice(fruits))
````

### `shuffle()`

`random.shuffle()` randomly rearranges the elements of a mutable sequence such as a list.

It changes the original list.


````python
import random

numbers = [1, 2, 3, 4, 5]

random.shuffle(numbers)

print(numbers)
````

## Trigonometric Functions

Python's `math` module provides trigonometric functions such as:

- `sin()`
- `cos()`
- `tan()`
- `asin()`
- `acos()`
- `atan()`

### Important

Python's trigonometric functions use **radians**, not degrees.

The functions `radians()` and `degrees()` can be used for conversion.


### `sin()`, `cos()` and `tan()`

These functions calculate the sine, cosine, and tangent of an angle measured in radians.


````python
import math

angle = math.pi / 2

print(math.sin(angle))
print(math.cos(angle))
print(math.tan(angle))
````

### Converting Degrees to Radians

Use `math.radians()` when the angle is given in degrees.


````python
import math

angle_degrees = 90
angle_radians = math.radians(angle_degrees)

print(angle_radians)
print(math.sin(angle_radians))
````

### Converting Radians to Degrees

Use `math.degrees()` when you need to convert radians back to degrees.


````python
import math

angle_radians = math.pi

print(math.degrees(angle_radians))
````

### Inverse Trigonometric Functions

The inverse functions are:

- `asin()` → inverse sine
- `acos()` → inverse cosine
- `atan()` → inverse tangent

Their results are returned in **radians**.


````python
import math

print(math.asin(1))
print(math.acos(1))
print(math.atan(1))

print(math.degrees(math.asin(1)))
print(math.degrees(math.atan(1)))
````

````python
import math

print(math.inf)
print(-math.inf)
````

# Important Interview Points

1. **`int`, `float`, and `complex` are Python's main built-in numeric types.**
2. `int` represents whole numbers, while `float` represents floating-point numbers.
3. A complex number is written using `j`, for example `3 + 4j`.
4. `int(10.8)` becomes `10`; converting a float to an integer truncates the decimal portion rather than rounding it.
5. `math.ceil()` rounds upward toward positive infinity, while `math.floor()` rounds downward toward negative infinity.
6. `random.randint(a, b)` includes both endpoints.
7. `random.randrange(start, stop)` excludes the stop value.
8. Python's trigonometric functions use **radians**.
9. Use `math.radians()` to convert degrees to radians and `math.degrees()` to convert radians to degrees.
10. `math.pi` represents π, `math.e` represents Euler's number, and `math.tau` represents `2π`.
