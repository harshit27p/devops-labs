# Python Data Types – Dictionary

A **dictionary** is a collection of **key-value pairs** in Python.

Each item in a dictionary consists of:

```text
key → value
```

Dictionaries are **mutable**, meaning their contents can be changed after the dictionary is created.

Dictionaries are commonly used when data needs to be stored and accessed using a meaningful **key** instead of a numeric index.

For example:

```text
"name" → "Shrishti"
"age"  → 23
"city" → "Bengaluru"
```


## Creating a Dictionary

A dictionary is created using curly braces `{}` with key-value pairs separated by commas.

### Syntax

```python
dictionary_name = {
    key1: value1,
    key2: value2
}
```

Each key is separated from its value using a colon `:`.

Example:


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "course": "Python"
}

print(student)
````

### Creating an Empty Dictionary

An empty dictionary can be created using `{}` or the `dict()` constructor.


````python
dict1 = {}
dict2 = dict()

print(dict1)
print(dict2)
````

### Dictionary with Different Data Types

Dictionary values can be of different data types. They can also contain lists, tuples, or other dictionaries.

Keys and values do not have to be of the same data type.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "marks": 91.5,
    "subjects": ["Python", "SQL", "ML"]
}

print(student)
````

## Accessing Values in Dictionary

Dictionary values are accessed using their **keys**.

Unlike lists and tuples, dictionaries are not accessed using positions such as `0`, `1`, `2`.

### Syntax

```python
dictionary_name[key]
```


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "course": "Python"
}

print(student["name"])
print(student["age"])
````

### Using `get()`

The `get()` method can also be used to access a value.

One important difference is that `get()` returns `None` if the key does not exist, instead of immediately raising a `KeyError`.

You can also provide a default value.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

print(student.get("name"))
print(student.get("city"))
print(student.get("city", "Not available"))
````

## Updating Dictionary

Dictionaries are **mutable**, so existing values can be changed and new key-value pairs can be added.

### Updating an Existing Value

Assign a new value to an existing key.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

student["age"] = 24

print(student)
````

### Adding a New Key-Value Pair

If the key does not already exist, assigning a value creates a new key-value pair.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

student["city"] = "Bengaluru"

print(student)
````

### Updating Multiple Values Using `update()`

The `update()` method can add new key-value pairs and update existing keys.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

student.update({
    "age": 24,
    "city": "Bengaluru"
})

print(student)
````

## Deleting Dictionary Elements

Dictionary elements can be deleted using:

- `del`
- `pop()`
- `popitem()`
- `clear()`



### Using `del`

`del` can be used to remove a specific key-value pair.

It can also be used to delete the entire dictionary.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "city": "Bengaluru"
}

del student["age"]

print(student)
````

### Using `pop()`

`pop()` removes the specified key and **returns its value**.

A default value can also be supplied if the key might not exist.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "city": "Bengaluru"
}

removed_value = student.pop("age")

print("Removed value:", removed_value)
print(student)
````

### Using `popitem()`

`popitem()` removes and returns the **last inserted key-value pair**.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "city": "Bengaluru"
}

removed_item = student.popitem()

print("Removed item:", removed_item)
print(student)
````

### Using `clear()`

`clear()` removes all key-value pairs from the dictionary, leaving an empty dictionary.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

student.clear()

print(student)
````

### Deleting the Entire Dictionary

The `del` statement can delete the dictionary variable itself.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

del student

# The dictionary no longer exists.
````

## Properties of Dictionary Keys

Dictionary keys have some important rules.

### 1. Keys must be unique

A dictionary cannot have two separate values stored under the same key.

If a key is repeated, the later value replaces the earlier value.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "age": 24
}

print(student)
````

### 2. Keys must be immutable / hashable

Dictionary keys must be hashable.

Common valid key types include:

- Strings
- Integers
- Floats
- Tuples containing hashable elements

Lists, dictionaries, and sets cannot be used directly as dictionary keys because they are unhashable.


````python
# Valid keys
data = {
    "name": "Shrishti",
    101: "Student ID",
    (1, 2): "Tuple key"
}

print(data)
````

### 3. Keys are case-sensitive

`"name"` and `"Name"` are treated as different keys.


````python
student = {
    "name": "Shrishti",
    "Name": "Different value"
}

print(student["name"])
print(student["Name"])
````

### 4. Dictionary keys are unique

Values can be duplicated, but keys must be unique.


````python
data = {
    "student1": "Python",
    "student2": "Python",
    "student3": "Python"
}

print(data)
````

## Built-in Dictionary Functions

Python provides several built-in functions that can be used with dictionaries.


### `len()`

Returns the number of key-value pairs in a dictionary.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "city": "Bengaluru"
}

print(len(student))
````

### `str()`

Converts a dictionary into its string representation.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

text = str(student)

print(text)
print(type(text))
````

### `type()`

Returns the data type of an object.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

print(type(student))
````

## Built-in Dictionary Methods

Dictionary methods are operations provided by the dictionary object.

The most important methods are:

- `keys()`
- `values()`
- `items()`
- `get()`
- `update()`
- `pop()`
- `popitem()`
- `clear()`
- `copy()`
- `setdefault()`


### `keys()`

Returns a view containing all the keys in the dictionary.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "city": "Bengaluru"
}

print(student.keys())
````

### `values()`

Returns a view containing all the values in the dictionary.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "city": "Bengaluru"
}

print(student.values())
````

### `items()`

Returns a view containing the dictionary's key-value pairs as tuples.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

print(student.items())
````

### Using `keys()`, `values()` and `items()` in a Loop

These methods are especially useful when iterating through a dictionary.


````python
student = {
    "name": "Shrishti",
    "age": 23,
    "city": "Bengaluru"
}

for key, value in student.items():
    print(key, ":", value)
````

### `get()`

Returns the value associated with a specified key.

If the key does not exist, `get()` returns `None` by default.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

print(student.get("name"))
print(student.get("city"))
````

### `update()`

Adds new key-value pairs and updates existing keys.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

student.update({"age": 24, "city": "Bengaluru"})

print(student)
````

### `copy()`

Creates a **shallow copy** of the dictionary.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

student_copy = student.copy()

print(student_copy)
````

### `setdefault()`

Returns the value of a key if it exists.

If the key does not exist, `setdefault()` adds the key with the specified default value and returns that value.


````python
student = {
    "name": "Shrishti"
}

age = student.setdefault("age", 23)

print(age)
print(student)
````

# Important Interview Points

1. **A dictionary stores data as key-value pairs.**
2. Dictionary keys must be **unique**.
3. Dictionary keys must be **hashable**.
4. Values can be duplicated and can be of different data types.
5. Dictionaries are **mutable**, so values can be added, changed, and removed.
6. `dict[key]` raises a `KeyError` if the key does not exist.
7. `get()` returns `None` by default when a key does not exist.
8. `remove()` is not a dictionary method; use `pop()` or `del` to remove a key-value pair.
9. `pop()` returns the removed value.
10. `popitem()` removes and returns the last inserted key-value pair.
11. `keys()`, `values()`, and `items()` return view objects.
12. Dictionary keys are case-sensitive.
