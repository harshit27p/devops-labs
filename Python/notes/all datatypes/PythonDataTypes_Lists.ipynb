# Python Data Types – Lists

A **list** is an ordered collection of items in Python. Lists are **mutable**, which means their elements can be changed after the list is created.

A list can contain elements of different data types, including numbers, strings, booleans, and even other lists.

Lists are created using square brackets `[]`, with individual elements separated by commas.

## Creating Lists

A list is created by placing elements inside square brackets `[]`.

### Syntax

```python
list_name = [element1, element2, element3]
```

A list can contain elements of the same or different data types.

````python
# List containing numbers
numbers = [10, 20, 30, 40]

# List containing strings
fruits = ["apple", "banana", "mango"]

# List containing different data types
mixed_list = [10, "Python", 3.14, True]

print(numbers)
print(fruits)
print(mixed_list)
````

### Creating an Empty List

An empty list contains no elements. It can be created using `[]` or the `list()` constructor.

````python
list1 = []
list2 = list()

print(list1)
print(list2)
````

## Accessing the Values in Lists

List elements can be accessed using their **index**.

Python uses **zero-based indexing**, so the first element is at index `0`.

````python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[0])
print(fruits[2])
print(fruits[3])
````

### Negative Indexing

Python also supports negative indexing. Negative indexes count from the end of the list.

```text
apple   banana   mango   orange
  0        1       2        3
 -4       -3      -2       -1
```

Therefore, `-1` refers to the last element.

````python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[-1])
print(fruits[-2])
````

## Updating Lists

Unlike strings, lists are **mutable**. This means individual elements can be changed after the list has been created.

To update an element, assign a new value to its index.

````python
fruits = ["apple", "banana", "mango"]

fruits[1] = "orange"

print(fruits)
````

### Updating Multiple Elements Using Slicing

A slice can also be used to replace multiple elements at once.

````python
numbers = [10, 20, 30, 40, 50]

numbers[1:3] = [200, 300]

print(numbers)
````

## Adding Elements to Lists

New elements can be added using methods such as `append()`, `insert()`, and `extend()`.

- `append()` adds one element at the end.
- `insert()` adds an element at a specific position.
- `extend()` adds multiple elements to the end.

````python
fruits = ["apple", "banana"]

fruits.append("mango")
print(fruits)

fruits.insert(1, "orange")
print(fruits)

fruits.extend(["grapes", "kiwi"])
print(fruits)
````

## Deleting List Elements

List elements can be deleted using `del`, `remove()`, `pop()`, or `clear()`.

- `del` removes an element using its index or can delete a slice.
- `remove()` removes the first matching value.
- `pop()` removes and returns an element, usually by index.
- `clear()` removes all elements from the list.

````python
# Using del
numbers = [10, 20, 30, 40]
del numbers[1]
print(numbers)
````

````python
# Using remove()
fruits = ["apple", "banana", "mango", "banana"]
fruits.remove("banana")
print(fruits)
````

````python
# Using pop()
fruits = ["apple", "banana", "mango"]

removed_item = fruits.pop(1)

print("Removed:", removed_item)
print(fruits)
````

````python
# Using clear()
numbers = [10, 20, 30]
numbers.clear()

print(numbers)
````

## Basic List Operators

Python provides several operators that can be used with lists.

### `+` Operator – Concatenation

The `+` operator combines two lists into a new list.

````python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

result = list1 + list2

print(result)
````

### `*` Operator – Repetition

The `*` operator repeats the elements of a list a specified number of times.

````python
numbers = [1, 2, 3]

print(numbers * 3)
````

### `in` and `not in` – Membership Operators

`in` checks whether an element exists in the list.

`not in` checks whether an element does not exist in the list.

````python
fruits = ["apple", "banana", "mango"]

print("banana" in fruits)
print("orange" not in fruits)
````

## Indexing, Slicing and Matrices

### Indexing

Indexing is used to access a single element from a list.

The first element has index `0`.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[3])
print(numbers[-1])
````

### Slicing

Slicing extracts a portion of a list.

### Syntax

```python
list[start:end:step]
```

The `start` index is included, while the `end` index is excluded.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])
````

### Lists as Matrices / Nested Lists

A list can contain other lists. This creates a **nested list**, which can be used to represent a matrix or table.

For example:

```text
1  2  3
4  5  6
7  8  9
```

This can be represented as:

```python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
```

The first index selects the row, and the second index selects the column.

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0][1])
print(matrix[2][0])
````

## Built-in List Functions

Python provides several built-in functions that work with lists.

### `len()`

Returns the number of elements in a list.

````python
numbers = [10, 20, 30, 40]

print(len(numbers))
````

### `max()` and `min()`

`max()` returns the largest element, while `min()` returns the smallest element.

````python
numbers = [10, 50, 20, 40, 30]

print(max(numbers))
print(min(numbers))
````

### `sum()`

Returns the sum of numeric elements in a list.

````python
numbers = [10, 20, 30, 40]

print(sum(numbers))
````

### `sorted()`

Returns a **new sorted list** without changing the original list.

````python
numbers = [40, 10, 30, 20]

sorted_numbers = sorted(numbers)

print(sorted_numbers)
print(numbers)
````

## Built-in List Methods

List methods are operations provided by the list object itself.

### `append()`

Adds one element to the end of the list.

````python
numbers = [1, 2, 3]
numbers.append(4)

print(numbers)
````

### `extend()`

Adds all elements from another iterable to the end of the list.

````python
numbers = [1, 2, 3]
numbers.extend([4, 5, 6])

print(numbers)
````

### `insert()`

Adds an element at a specified index.

````python
numbers = [1, 2, 4]
numbers.insert(2, 3)

print(numbers)
````

### `remove()`

Removes the first occurrence of a specified value.

If the value is not present, `remove()` raises a `ValueError`.

````python
fruits = ["apple", "banana", "mango", "banana"]
fruits.remove("banana")

print(fruits)
````

### `pop()`

Removes and returns an element from the list. If no index is provided, it removes the last element.

````python
numbers = [10, 20, 30, 40]

last_item = numbers.pop()
print("Removed:", last_item)
print(numbers)
````

### `index()`

Returns the index of the first occurrence of a specified value.

````python
fruits = ["apple", "banana", "mango"]

print(fruits.index("mango"))
````

### `count()`

Returns the number of times a value appears in the list.

````python
numbers = [1, 2, 2, 3, 2, 4]

print(numbers.count(2))
````

### `sort()`

Sorts the list **in place**, meaning the original list is changed.

````python
numbers = [40, 10, 30, 20]

numbers.sort()

print(numbers)
````

### `reverse()`

Reverses the order of elements in the original list.

````python
numbers = [1, 2, 3, 4]

numbers.reverse()

print(numbers)
````

### `clear()`

Removes all elements from the list, leaving an empty list.

````python
numbers = [1, 2, 3]
numbers.clear()

print(numbers)
````

4. In slicing, the start index is included and the end index is excluded.
