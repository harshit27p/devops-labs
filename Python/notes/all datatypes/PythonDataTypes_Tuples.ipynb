# Python Data Types – Tuples

A **tuple** is an ordered collection of elements in Python.

Tuples are similar to lists, but there is one major difference:

> **Tuples are immutable**, which means their elements cannot be changed after the tuple is created.

Tuples are created using parentheses `()` and can contain elements of different data types.


## Creating the Tuple

A tuple is usually created by placing elements inside parentheses `()` and separating them with commas.

### Syntax

```python
tuple_name = (element1, element2, element3)
```

A tuple can contain elements of the same or different data types.


````python
# Tuple containing numbers
numbers = (10, 20, 30, 40)

# Tuple containing strings
fruits = ("apple", "banana", "mango")

# Tuple containing different data types
mixed_tuple = (10, "Python", 3.14, True)

print(numbers)
print(fruits)
print(mixed_tuple)
````

### Creating an Empty Tuple

An empty tuple can be created using `()` or the `tuple()` constructor.


````python
tuple1 = ()
tuple2 = tuple()

print(tuple1)
print(tuple2)
````

### Creating a Tuple with One Element

A single-element tuple requires a **comma**.

Without the comma, Python treats it as a normal value inside parentheses.


````python
# Correct: single-element tuple
tuple1 = (10,)

# Not a tuple
value = (10)

print(tuple1)
print(type(tuple1))
print(value)
print(type(value))
````

## Accessing Values in Tuples

Tuple elements can be accessed using their **index**.

Python uses **zero-based indexing**, so the first element is at index `0`.


````python
fruits = ("apple", "banana", "mango", "orange")

print(fruits[0])
print(fruits[2])
print(fruits[3])
````

### Negative Indexing

Tuples also support negative indexing. Negative indexes count from the end of the tuple.

```text
apple   banana   mango   orange
  0        1       2        3
 -4       -3      -2       -1
```

Therefore, `-1` refers to the last element.


````python
fruits = ("apple", "banana", "mango", "orange")

print(fruits[-1])
print(fruits[-2])
````

## Updating Tuples

Tuples are **immutable**, so individual elements cannot be changed after the tuple has been created.

For example, the following will produce a `TypeError`:

```python
fruits[1] = "orange"
```

Unlike lists, tuples do not allow item assignment.

If you need a modified tuple, you can create a **new tuple**.


````python
fruits = ("apple", "banana", "mango")

# Create a new tuple instead of changing the original
new_fruits = fruits[:1] + ("orange",) + fruits[2:]

print(fruits)
print(new_fruits)
````

### Important

Although tuple elements cannot be changed, a tuple can contain a **mutable object** such as a list. That inner list can still be modified.

The tuple itself remains unchanged in terms of its references.


````python
data = ("Python", [10, 20, 30])

data[1].append(40)

print(data)
````

## Delete Tuple Elements

Because tuples are immutable, individual elements cannot be deleted.

For example:

```python
del fruits[1]
```

will produce a `TypeError`.

However, you can delete the **entire tuple** using the `del` statement.


````python
fruits = ("apple", "banana", "mango")

del fruits

# The tuple no longer exists.
````

## Basic Tuple Operations

Python provides several operators that can be used with tuples.

The most common operations are:

- `+` → Concatenation
- `*` → Repetition
- `in` → Membership
- `not in` → Non-membership


### `+` Operator – Concatenation

The `+` operator combines two tuples into a new tuple.


````python
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

result = tuple1 + tuple2

print(result)
````

### `*` Operator – Repetition

The `*` operator repeats the elements of a tuple a specified number of times.


````python
numbers = (1, 2, 3)

print(numbers * 3)
````

### `in` and `not in` – Membership Operators

`in` checks whether an element exists in the tuple.

`not in` checks whether an element does not exist in the tuple.


````python
fruits = ("apple", "banana", "mango")

print("banana" in fruits)
print("orange" not in fruits)
````

## Indexing, Slicing and Matrices

### Indexing

Indexing is used to access a single element from a tuple.

The first element has index `0`.


````python
numbers = (10, 20, 30, 40, 50)

print(numbers[0])
print(numbers[3])
print(numbers[-1])
````

### Slicing

Slicing extracts a portion of a tuple.

### Syntax

```python
tuple[start:end:step]
```

The `start` index is included, while the `end` index is excluded.


````python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])
````

### Tuples as Matrices / Nested Tuples

A tuple can contain other tuples. This creates a **nested tuple**, which can be used to represent a matrix or table.

For example:

```text
1  2  3
4  5  6
7  8  9
```

This can be represented as:

```python
matrix = ((1, 2, 3),
          (4, 5, 6),
          (7, 8, 9))
```

The first index selects the row, and the second index selects the column.


````python
matrix = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
)

print(matrix[0][1])
print(matrix[2][0])
````

## Built-in Tuple Functions

Python provides several built-in functions that can be used with tuples.


### `len()`

Returns the number of elements in a tuple.


````python
numbers = (10, 20, 30, 40)

print(len(numbers))
````

### `max()` and `min()`

`max()` returns the largest element, while `min()` returns the smallest element.


````python
numbers = (10, 50, 20, 40, 30)

print(max(numbers))
print(min(numbers))
````

### `sum()`

Returns the sum of numeric elements in a tuple.


````python
numbers = (10, 20, 30, 40)

print(sum(numbers))
````

### `sorted()`

Returns a **new sorted list** containing the elements of the tuple.

Notice that the result is a **list**, not a tuple.


````python
numbers = (40, 10, 30, 20)

sorted_numbers = sorted(numbers)

print(sorted_numbers)
print(type(sorted_numbers))
````

### `tuple()`

The `tuple()` function can be used to convert another iterable, such as a list, into a tuple.


````python
numbers_list = [10, 20, 30, 40]

numbers_tuple = tuple(numbers_list)

print(numbers_tuple)
print(type(numbers_tuple))
````

## Built-in Tuple Methods

Tuples have only a small number of methods because they are immutable.

The two main tuple methods are:

- `count()`
- `index()`


### `count()`

Returns the number of times a specified value appears in the tuple.


````python
numbers = (1, 2, 2, 3, 2, 4)

print(numbers.count(2))
````

### `index()`

Returns the index of the first occurrence of a specified value.


````python
fruits = ("apple", "banana", "mango")

print(fruits.index("mango"))
````
