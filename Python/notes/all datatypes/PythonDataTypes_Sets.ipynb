# Python Data Types – Sets

A **set** is an unordered collection of **unique elements** in Python.

Sets are useful when you need to:
- Store only unique values
- Perform fast membership testing
- Perform mathematical operations such as union, intersection, and difference

Sets are **mutable**, which means their contents can be changed after creation.


## Creating A Set

A set can be created by placing elements inside curly braces `{}`.

### Syntax

```python
set_name = {element1, element2, element3}
```

Duplicate values are automatically removed because a set stores only unique elements.


````python
numbers = {10, 20, 30, 40}

print(numbers)
````

### Creating a Set with Duplicate Values

If the same value appears more than once, the set keeps only one copy.


````python
numbers = {10, 20, 20, 30, 30, 40}

print(numbers)
````

### Creating an Empty Set

Be careful:

```python
{}
```

creates an **empty dictionary**, not an empty set.

Use `set()` to create an empty set.


````python
empty_set = set()
empty_dictionary = {}

print(empty_set)
print(type(empty_set))

print(empty_dictionary)
print(type(empty_dictionary))
````

### Creating a Set from Another Iterable

The `set()` constructor can convert a list, tuple, or string into a set. Duplicate elements are removed.


````python
numbers_list = [10, 20, 20, 30, 30]

numbers_set = set(numbers_list)

print(numbers_set)
````

## Accessing A Set

Sets are **unordered**, so they do not support indexing or slicing like lists and tuples.

Instead, you can:
- Check membership using `in`
- Iterate through the elements using a `for` loop


````python
fruits = {"apple", "banana", "mango"}

print("banana" in fruits)
print("orange" in fruits)
````

### Iterating Through a Set

A `for` loop can be used to access each element.

Because sets are unordered, do not depend on a particular iteration order.


````python
fruits = {"apple", "banana", "mango"}

for fruit in fruits:
    print(fruit)
````

## Modifying A Set

Sets are mutable, so elements can be added and removed after the set is created.

Common methods for modifying sets include `add()` and `update()`.


### `add()`

`add()` adds a **single element** to the set.

If the element already exists, the set remains unchanged.


````python
fruits = {"apple", "banana"}

fruits.add("mango")
print(fruits)

fruits.add("apple")  # No duplicate is created
print(fruits)
````

### `update()`

`update()` adds **multiple elements** from another iterable such as a list, tuple, or another set.


````python
fruits = {"apple", "banana"}

fruits.update(["mango", "orange", "grapes"])

print(fruits)
````

## Removing Items From A Set

Set elements can be removed using:
- `remove()`
- `discard()`
- `pop()`
- `clear()`


### `remove()`

`remove()` removes a specified element.

If the element does not exist, `remove()` raises a `KeyError`.


````python
fruits = {"apple", "banana", "mango"}

fruits.remove("banana")

print(fruits)
````

### `discard()`

`discard()` also removes a specified element.

The difference is that `discard()` does **not** raise an error if the element is absent.


````python
fruits = {"apple", "banana", "mango"}

fruits.discard("banana")
fruits.discard("orange")  # No error

print(fruits)
````

### `pop()`

`pop()` removes and returns an **arbitrary element** from the set.

Because sets are unordered, you should not assume which element will be removed.


````python
fruits = {"apple", "banana", "mango"}

removed_item = fruits.pop()

print("Removed:", removed_item)
print(fruits)
````

### `clear()`

`clear()` removes all elements from the set, leaving an empty set.


````python
numbers = {10, 20, 30}

numbers.clear()

print(numbers)
````

## Common Set Operations

The main mathematical set operations are:
- Union
- Intersection
- Difference
- Symmetric Difference


### Union

The **union** contains all unique elements from both sets.

It can be performed using `|` or `union()`.


````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(set_a | set_b)
print(set_a.union(set_b))
````

### Intersection

The **intersection** contains only elements common to both sets.

It can be performed using `&` or `intersection()`.


````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(set_a & set_b)
print(set_a.intersection(set_b))
````

### Difference

The **difference** contains elements present in the first set but not in the second set.

It can be performed using `-` or `difference()`.

Note that `A - B` and `B - A` can produce different results.


````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(set_a - set_b)
print(set_b - set_a)
````

### Symmetric Difference

The **symmetric difference** contains elements that are in either set, but **not in both**.

It can be performed using `^` or `symmetric_difference()`.


````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(set_a ^ set_b)
print(set_a.symmetric_difference(set_b))
````

### Set Operations at a Glance

For two sets `A` and `B`:

```text
A | B  → Union
A & B  → Intersection
A - B  → Difference
A ^ B  → Symmetric Difference
```


## Subset and Superset Operations

Sets can also be compared to determine whether one set is contained within another.

### `issubset()`

Checks whether all elements of one set are present in another set.


````python
small_set = {1, 2}
large_set = {1, 2, 3, 4}

print(small_set.issubset(large_set))
````

### `issuperset()`

Checks whether a set contains all elements of another set.


````python
small_set = {1, 2}
large_set = {1, 2, 3, 4}

print(large_set.issuperset(small_set))
````

### `isdisjoint()`

Checks whether two sets have **no elements in common**.


````python
set_a = {1, 2, 3}
set_b = {4, 5, 6}

print(set_a.isdisjoint(set_b))
````

## Built-in Set Functions

Common built-in functions that work with sets include:
- `len()`
- `max()`
- `min()`
- `sum()`
- `sorted()`


### `len()`

Returns the number of elements in a set.


````python
numbers = {10, 20, 30, 40}

print(len(numbers))
````

### `max()` and `min()`

`max()` returns the largest element, while `min()` returns the smallest element.


````python
numbers = {10, 50, 20, 40, 30}

print(max(numbers))
print(min(numbers))
````

### `sum()`

Returns the sum of numeric elements in a set.


````python
numbers = {10, 20, 30, 40}

print(sum(numbers))
````

### `sorted()`

`sorted()` returns a **new sorted list** containing the elements of the set.

Remember:

> A set is unordered, but `sorted()` produces a list in sorted order.


````python
numbers = {40, 10, 30, 20}

sorted_numbers = sorted(numbers)

print(sorted_numbers)
print(type(sorted_numbers))
````

## Built-in Set Methods

Important set methods include:

- `add()`
- `update()`
- `remove()`
- `discard()`
- `pop()`
- `clear()`
- `union()`
- `intersection()`
- `difference()`
- `symmetric_difference()`
- `issubset()`

- `copy()`


### `copy()`

`copy()` creates a shallow copy of a set.


````python
numbers = {10, 20, 30}

numbers_copy = numbers.copy()

print(numbers_copy)
````

### `intersection_update()`

`intersection_update()` keeps only the elements common to the set and another set.

It **modifies the original set**.


````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

set_a.intersection_update(set_b)

print(set_a)
````

### `difference_update()`

`difference_update()` removes elements that are also present in another set.

It **modifies the original set**.


````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

set_a.difference_update(set_b)

print(set_a)
````

### `symmetric_difference_update()`

`symmetric_difference_update()` updates the original set so that it contains the symmetric difference between the two sets.


````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

set_a.symmetric_difference_update(set_b)

print(set_a)
````

# Important Interview Points

1. **Sets contain unique elements**, so duplicate values are automatically removed.
2. Sets are **unordered**, so they do not support indexing or slicing.
3. Use `in` to check whether an element exists in a set.
4. `{}` creates an empty **dictionary**, while `set()` creates an empty set.
5. `remove()` raises a `KeyError` if the element does not exist; `discard()` does not.
6. `pop()` removes and returns an arbitrary element because sets are unordered.
7. `|`, `&`, `-`, and `^` represent union, intersection, difference, and symmetric difference.
8. `sorted()` can be used with a set, but it returns a **list**, not a set.
9. Set elements must be **hashable**, so mutable objects such as lists and dictionaries cannot be direct set elements.
10. Sets are useful for removing duplicates and performing mathematical set operations.
