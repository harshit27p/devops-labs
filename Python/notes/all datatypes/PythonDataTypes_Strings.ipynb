## Accessing Values in Strings

A string is a sequence of characters, so each character has a position called an index.

```python
name = "PYTHON"
```

Its indexes are:

 P   Y   T   H   O   N


 0   1   2   3   4   5

Python uses **zero-based indexing**, meaning the first character is at index 0.

Accessing individual characters

```python
name = "PYTHON"

print(name[0])   # P
print(name[2])   # T
print(name[5])   # N
```

Negative Indexing

Python also allows negative indexes

 P    Y    T    H    O    N


-6   -5   -4   -3   -2   -1

```python
name = "PYTHON"

print(name[-1])   # N
print(name[-2])   # O
print(name[-6])   # P
```

3.String Slicing

Slicing means extracting a portion of a string.

Syntax:

string[start:end]

Important:

**start is included, but end is excluded**.

```python
name = "PYTHON"

print(name[0:3])
```

```python
name = "PYTHON"

print(name[1:4])   # YTH
print(name[2:5])   # THO
print(name[:3])    # PYT
print(name[3:])    # HON
print(name[:])     # PYTHON
```

With step

Syntax:

string[start:end:step]

```python
name = "PYTHON"

print(name[::2])
```

Reverse a string:

```python
print(name[::-1])
```

## Updating Strings

Python strings are immutable.

Immutable means that once a string is created, you cannot directly change an individual character.

```python
name = "Python"

name[0] = "J"
```

So how do we "update" a string?

We create a new string.

```python
name = "Python"

name = "J" + name[1:]

print(name)
```

```python
name = "Python"

name = name.replace("Python", "Java")

print(name)
```

The original string wasn't modified. A new string was created and assigned to name.

# String Formatting Operator

The % operator is an older way of formatting strings.

It allows you to insert values into a string.

%s — String

```python
name = "Shrishti"

print("My name is %s" % name)
```

%d — Integer

```python
age = 23

print("I am %d years old" % age)
```

%f — Floating-point number

```python
price = 99.5

print("Price = %f" % price)
```

You can control decimal places:

```python
print("Price = %.2f" % price)
```

multiple values:

```python
name = "Shrishti"
age = 23

print("My name is %s and I am %d years old." % (name, age))
```

Although % formatting is part of Python, f-strings are usually preferred in modern Python.

```python
name = "Shrishti"
age = 23

print(f"My name is {name} and I am {age} years old.")
```

You can also put expressions inside {}:

```python
a = 10
b = 20

print(f"Sum = {a + b}")
```

For decimal formatting:

```python
price = 99.5678

print(f"Price: ₹{price:.2f}")
```

# Built-in String Methods

Python provides many built-in methods for working with strings.

A method is a function associated with an object.

upper()

Converts all letters to uppercase.

```python
text = "hello python"

print(text.upper())
```

lower()

Converts all letters to lowercase.

```python
text = "HELLO PYTHON"

print(text.lower())
```

capitalize()

Makes the first character uppercase and the remaining characters lowercase.

```python
text = "hello PYTHON"

print(text.capitalize())
```

title()

Capitalizes the first letter of each word

```python
text = "python programming language"

print(text.title())
```

strip(), lstrip(), rstrip()

These are extremely useful when processing user input.

strip()

Removes spaces from both ends.

```python
text = "   Hello Python   "

print(text.strip())
```

lstrip()

Removes spaces from the left.

rstrip()

Removes spaces from the right.

```python
text = "   Hello"

print(text.lstrip())
```

Important

These methods don't remove spaces between words.

replace()

Replaces one substring with another.

Syntax:

string.replace(old, new)

```python
text = "I like Java"

new_text = text.replace("Java", "Python")

print(new_text)
```

You can also specify how many replacements:

```python
text = "apple apple apple"

print(text.replace("apple", "mango", 2))
```

split()

Splits a string into a list.

```python
text = "Python is easy"

words = text.split()

print(words)
```

By default, it splits based on whitespace.

You can specify a separator:

```python
data = "apple,banana,mango"

fruits = data.split(",")

print(fruits)
```

join()

join() does the opposite of split().

It combines elements of a sequence into a string.

```python
words = ["Python", "is", "easy"]

sentence = " ".join(words)

print(sentence)
```

```python
fruits = ["apple", "banana", "mango"]

result = ", ".join(fruits)

print(result)
```

find()

Finds the position of the first occurrence of a substring.

```python
text = "Python programming"

print(text.find("program"))
```

Because "program" starts at index 7.

If the substring isn't found:

```python
print(text.find("Java"))
```

index()

Similar to find().

```python
text = "Python programming"

print(text.index("program"))
```

Difference between find() and index()

If the substring doesn't exist:

text.find("Java")

returns:

-1

But:

text.index("Java")

raises:

ValueError

So find() is safer when you're not sure whether something exists.

count()

Counts how many times a substring occurs.

```python
text = "banana"

print(text.count("a"))
```

```python
text = "hello hello"

print(text.count("hello"))
```
