# Decision Making in Python

Decision-making statements allow a program to **choose which block of code to execute based on a condition**.

Python mainly provides:

- `if` statement
- `if-else` statement
- `if-elif-else` statement
- Nested `if` statements

These statements are based on conditions that evaluate to either `True` or `False`.


## Single Statement Suites

A **suite** is a group of statements that belongs to a control statement such as `if`.

When the suite contains only **one statement**, Python allows it to be written on the same line as the `if` statement.

### Syntax

```python
if condition: statement
```

This is useful for very short and simple conditions.


````python
age = 20

if age >= 18: print("Eligible to vote")
````

### Single Statement with `if-else`

A simple `if-else` can also be written on a single line.

```python
if condition: statement1
else: statement2
```

However, for complex logic, writing the statements on separate indented lines is easier to read.


````python
age = 16

if age >= 18: print("Adult")
else: print("Minor")
````

## `if` Statements

The `if` statement executes a block of code **only when its condition is `True`**.

### Syntax

```python
if condition:
    statement(s)
```

The colon `:` marks the beginning of the `if` suite, and indentation tells Python which statements belong to it.


````python
age = 20

if age >= 18:
    print("You are an adult")
````

### How an `if` Statement Works

1. Python evaluates the condition.
2. If the condition is `True`, the indented block executes.
3. If the condition is `False`, the block is skipped.
4. Execution then continues with the statement after the `if` block.


````python
marks = 75

if marks >= 40:
    print("Pass")

print("Program continues...")
````

### `if` with a Comparison

Conditions are often created using comparison operators such as:

- `>`
- `<`
- `==`
- `!=`
- `>=`
- `<=`


````python
number = 10

if number > 0:
    print("Positive number")
````

### `if` with Logical Operators

Multiple conditions can be combined using `and`, `or`, and `not`.


````python
age = 22
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
````

## `if else` Statements

The `if-else` statement provides **two possible paths**.

- If the `if` condition is `True`, the `if` block executes.
- Otherwise, the `else` block executes.

### Syntax

```python
if condition:
    statement(s)
else:
    statement(s)
```


````python
age = 16

if age >= 18:
    print("Adult")
else:
    print("Minor")
````

### Example: Even or Odd

The modulus operator `%` can be used to check whether a number is divisible by `2`.


````python
number = 17

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
````

### Example: Pass or Fail

A condition can be used to make a simple pass/fail decision.


````python
marks = 65

if marks >= 40:
    print("Pass")
else:
    print("Fail")
````

## `if elif` Statements

When there are **multiple conditions**, use `if-elif-else`.

Python checks the conditions from top to bottom.

- The first `True` condition gets executed.
- Once a condition is satisfied, the remaining `elif` and `else` blocks are skipped.
- The `else` block executes only when none of the conditions are `True`.

### Syntax

```python
if condition1:
    statement(s)
elif condition2:
    statement(s)
elif condition3:
    statement(s)
else:
    statement(s)
```


````python
marks = 82

if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
````

### Example: Checking Temperature

`if-elif-else` is useful when a value can fall into different ranges.


````python
temperature = 32

if temperature >= 40:
    print("Very hot")
elif temperature >= 30:
    print("Hot")
elif temperature >= 20:
    print("Moderate")
else:
    print("Cold")
````

### Important Point About `elif`

Only **one branch** of an `if-elif-else` chain is executed.

Python stops checking once it finds the first condition that evaluates to `True`.


````python
x = 15

if x > 10:
    print("Greater than 10")
elif x > 5:
    print("Greater than 5")
else:
    print("5 or less")
````

## Nested Statements

A **nested statement** means placing one decision-making statement inside another.

For example, an `if` statement can contain another `if` statement.

### Syntax

```python
if condition1:
    if condition2:
        statement(s)
    else:
        statement(s)
else:
    statement(s)
```

The inner condition is checked only when the outer condition is satisfied.


````python
age = 22
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Not eligible")
````

### Example: Login Validation

Nested statements can be useful when one decision depends on another decision.


````python
username = "admin"
password = "python123"

if username == "admin":
    if password == "python123":
        print("Login successful")
    else:
        print("Incorrect password")
else:
    print("Unknown username")
````

### Nested `if-elif-else`

A nested structure can contain an `if-elif-else` chain as well.


````python
age = 25
income = 60000

if age >= 18:
    if income >= 50000:
        print("Eligible for premium plan")
    elif income >= 30000:
        print("Eligible for standard plan")
    else:
        print("Eligible for basic plan")
else:
    print("Not eligible")
````

## Indentation in Decision Making

Python uses **indentation to define the block of code** belonging to an `if`, `elif`, or `else`.

The statements inside the block must have the same indentation level.


````python
marks = 80

if marks >= 40:
    print("Pass")
    print("Congratulations!")

print("This statement is outside the if block.")
````

### Incorrect Indentation

Incorrect indentation can cause an `IndentationError`.

For example:

```python
if marks >= 40:
print("Pass")
```

The `print()` statement must be indented because it belongs to the `if` block.


## Difference Between `if`, `if-else`, `if-elif`, and Nested `if`

| Statement | When to use |
|---|---|
| `if` | When you need to execute code only if one condition is true |
| `if-else` | When there are two possible paths |
| `if-elif-else` | When there are multiple mutually exclusive conditions |
| Nested `if` | When one decision depends on another decision |


## Quick Revision

### `if`

```python
if condition:
    statement
```

Executes the statement only when the condition is `True`.

### `if-else`

```python
if condition:
    statement1
else:
    statement2
```

Chooses between two paths.

### `if-elif-else`

```python
if condition1:
    statement1
elif condition2:
    statement2
else:
    statement3
```

Chooses one path from multiple conditions.

### Nested `if`

```python
if condition1:
    if condition2:
        statement
```

Places one decision inside another decision.


# Important Interview Points

1. Python uses `if`, `elif`, and `else` for decision making.
2. A colon `:` is required after the condition.
3. Python uses **indentation** to define the statements belonging to a decision-making block.
4. An `if` block executes only when its condition is `True`.
5. `else` executes when the preceding `if` condition is `False`.
6. In an `if-elif-else` chain, Python executes the **first matching condition** and skips the remaining branches.
7. `elif` means **else if**.
8. Multiple `elif` blocks can be used.
9. `else` is optional.
10. Nested `if` statements are useful when one decision depends on another.
11. A simple one-line `if` can be written as a **single statement suite**, but normal multi-line indentation is generally easier to read.
