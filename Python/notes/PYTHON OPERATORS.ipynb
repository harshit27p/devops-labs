# Python Operators

Operators are symbols or keywords used to perform operations on values and variables.

For example:

```python
a + b
```

Here:
- `a` and `b` are **operands**
- `+` is the **operator**

Python provides several types of operators for arithmetic, comparison, assignment, logical operations, bit manipulation, membership testing, and identity checking.


## List of Operators

The main types of Python operators are:

| Operator Type | Purpose |
|---|---|
| Arithmetic Operators | Perform mathematical calculations |
| Comparison Operators | Compare two values |
| Assignment Operators | Assign or update values |
| Bitwise Operators | Perform operations on individual bits |
| Logical Operators | Combine or modify conditions |
| Membership Operators | Check whether a value exists in a sequence |
| Identity Operators | Check whether two variables refer to the same object |
| Operator Precedence | Determines the order in which operators are evaluated |



## Arithmetic Operators

Arithmetic operators are used to perform mathematical calculations.

| Operator | Name | Example |
|---|---|---|
| `+` | Addition | `a + b` |
| `-` | Subtraction | `a - b` |
| `*` | Multiplication | `a * b` |
| `/` | Division | `a / b` |
| `//` | Floor Division | `a // b` |
| `%` | Modulus | `a % b` |
| `**` | Exponentiation | `a ** b` |


````python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
````

### `/` Division vs `//` Floor Division

`/` performs normal division and returns a floating-point result.

`//` performs floor division and returns the floor of the result.

The floor means the greatest integer less than or equal to the result.


````python
print(10 / 3)
print(10 // 3)

print(-10 / 3)
print(-10 // 3)
````

### `%` Modulus Operator

The `%` operator returns the **remainder** after division.

It is commonly used to check whether a number is divisible by another number.


````python
print(10 % 3)
print(20 % 5)

number = 17

print("Even:", number % 2 == 0)
````

### `**` Exponentiation Operator

The `**` operator raises the left operand to the power of the right operand.


````python
print(2 ** 3)
print(5 ** 2)
````

## Comparison Operators

Comparison operators compare two values and return either `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |


````python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
````

### `=` vs `==`

These two operators are very different:

- `=` → **assignment operator**
- `==` → **comparison operator**

```python
x = 10       # Assign 10 to x
x == 10      # Check whether x is equal to 10
```


````python
x = 10

print(x == 10)
print(x == 20)
````

## Assignment Operators

Assignment operators are used to assign values to variables.

The basic assignment operator is `=`.

Python also provides **compound assignment operators**, which perform an operation and then assign the result back to the variable.


### Basic Assignment

```python
variable = value
```


````python
x = 10

print(x)
````

### Compound Assignment Operators

| Operator | Equivalent to |
|---|---|
| `+=` | `x = x + value` |
| `-=` | `x = x - value` |
| `*=` | `x = x * value` |
| `/=` | `x = x / value` |
| `//=` | `x = x // value` |
| `%=` | `x = x % value` |
| `**=` | `x = x ** value` |
| `&=` | `x = x & value` |
| `|=` | `x = x | value` |
| `^=` | `x = x ^ value` |
| `>>=` | `x = x >> value` |
| `<<=` | `x = x << value` |


````python
x = 10

x += 5
print(x)

x -= 3
print(x)

x *= 2
print(x)

x //= 4
print(x)
````

## Bitwise Operators

Bitwise operators work on the **individual binary bits** of integer values.

The main bitwise operators are:

| Operator | Name |
|---|---|
| `&` | Bitwise AND |
| `|` | Bitwise OR |
| `^` | Bitwise XOR |
| `~` | Bitwise NOT |
| `<<` | Left Shift |
| `>>` | Right Shift |


### Understanding Binary Representation

For example:

```text
5  → 0101
3  → 0011
```

Bitwise operators compare or manipulate these individual bits.


### `&` Bitwise AND

A bit becomes `1` only when **both corresponding bits are `1`**.


````python
a = 5      # 0101
b = 3      # 0011

print(a & b)
````

### `|` Bitwise OR

A bit becomes `1` when **at least one of the corresponding bits is `1`**.


````python
a = 5      # 0101
b = 3      # 0011

print(a | b)
````

### `^` Bitwise XOR

A bit becomes `1` when the corresponding bits are **different**.


````python
a = 5      # 0101
b = 3      # 0011

print(a ^ b)
````

### `~` Bitwise NOT

`~` flips the bits of an integer.

In Python, integers use signed representation, so the result follows:

```text
~x = -(x + 1)
```


````python
a = 5

print(~a)
````

### `<<` Left Shift

The left-shift operator moves the bits to the left by the specified number of positions.

For non-negative integers, shifting left by one position is equivalent to multiplying by `2`.


````python
a = 5      # 0101

print(a << 1)
print(a << 2)
````

### `>>` Right Shift

The right-shift operator moves the bits to the right by the specified number of positions.

For non-negative integers, shifting right by one position is equivalent to integer division by `2`.


````python
a = 20

print(a >> 1)
print(a >> 2)
````

## Logical Operators

Logical operators are used to combine or modify conditions.

Python has three logical operators:

| Operator | Meaning |
|---|---|
| `and` | True if both conditions are true |
| `or` | True if at least one condition is true |
| `not` | Reverses the truth value |


### `and`

Returns a truthy result when **both conditions** are true.

It is commonly used when multiple conditions must be satisfied.


````python
age = 23
has_id = True

print(age >= 18 and has_id)
````

### `or`

Returns a truthy result when **at least one condition** is true.


````python
is_student = True
is_employee = False

print(is_student or is_employee)
````

### `not`

`not` reverses a Boolean condition.

- `not True` → `False`
- `not False` → `True`


````python
is_raining = False

print(not is_raining)
````

### Short-Circuit Evaluation

Python uses **short-circuit evaluation** with `and` and `or`.

For `and`, if the first operand is false, Python does not need to evaluate the remaining operands.

For `or`, if the first operand is true, Python does not need to evaluate the remaining operands.

This can improve efficiency and can also prevent unnecessary or unsafe evaluations.


````python
x = 0

print(x != 0 and 10 / x > 2)
````

## Membership Operators

Membership operators check whether a value exists in a sequence or collection.

Python provides:

| Operator | Meaning |
|---|---|
| `in` | Returns `True` if the value exists |
| `not in` | Returns `True` if the value does not exist |


### `in`

`in` checks whether an element exists in a collection.


````python
fruits = ["apple", "banana", "mango"]

print("banana" in fruits)
print("orange" in fruits)
````

### `not in`

`not in` checks whether an element does not exist in a collection.


````python
fruits = ["apple", "banana", "mango"]

print("orange" not in fruits)
print("banana" not in fruits)
````

### Membership with Strings

Membership operators can also check for a substring inside a string.


````python
text = "Python programming"

print("Python" in text)
print("Java" not in text)
````

### Membership with Dictionaries

When used directly with a dictionary, `in` checks the **keys**, not the values.


````python
student = {
    "name": "Shrishti",
    "age": 23
}

print("name" in student)
print("Shrishti" in student)
````

## Identity Operators

Identity operators check whether two variables refer to the **same object in memory**.

Python provides:

| Operator | Meaning |
|---|---|
| `is` | Same object |
| `is not` | Different objects |


### `is`

`is` checks object identity, not just whether two values are equal.

For value comparison, use `==`.

A useful example is checking against `None`:


````python
value = None

print(value is None)
print(value is not None)
````

### `==` vs `is`

This is an important interview distinction.

- `==` checks whether two objects have equal **values**.
- `is` checks whether two variables refer to the **same object**.


````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
````

Here, `a` and `b` contain the same values, so `a == b` is `True`.

However, they are two separate list objects, so `a is b` is `False`.

Therefore:

> Use `==` for value equality and `is` for object identity.


## Operators Precedence

When an expression contains multiple operators, Python needs to determine **which operation should be performed first**.

This is called **operator precedence**.

For example:

```python
result = 2 + 3 * 4
```

Multiplication has higher precedence than addition, so:

```text
2 + (3 * 4)
= 2 + 12
= 14
```


````python
result = 2 + 3 * 4

print(result)
````

### Common Operator Precedence Order

From **higher precedence to lower precedence**:

| Priority | Operators / Expression |
|---|---|
| 1 | `()` |
| 2 | `**` |
| 3 | `+x`, `-x`, `~x` |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `<<`, `>>` |
| 7 | `&` |
| 8 | `^` |
| 9 | `|` |
| 10 | Comparisons: `<`, `<=`, `>`, `>=`, `==`, `!=`, `is`, `is not`, `in`, `not in` |
| 11 | `not` |
| 12 | `and` |
| 13 | `or` |

Assignment operators such as `=`, `+=`, and `-=` have lower precedence than the operators above.


### Using Parentheses

Parentheses have the highest priority in normal expressions and can be used to explicitly control the order of evaluation.

Using parentheses also makes complex expressions easier to understand.


````python
result1 = 2 + 3 * 4
result2 = (2 + 3) * 4

print(result1)
print(result2)
````

### Example with Logical Operators

Logical operators also follow precedence rules.

`not` has higher precedence than `and`, and `and` has higher precedence than `or`.

So:

```python
True or False and False
```

is evaluated as:

```python
True or (False and False)
```


````python
result = True or False and False

print(result)
````

# Important Interview Points

1. `=` is an **assignment** operator, while `==` is a **comparison** operator.
2. `/` performs normal division, while `//` performs floor division.
3. `%` returns the remainder after division.
4. `**` is the exponentiation operator.
5. Comparison operators return Boolean results such as `True` or `False`.
6. `and`, `or`, and `not` are logical operators.
7. `in` and `not in` are membership operators.
8. `is` and `is not` check **object identity**, while `==` and `!=` check **value equality/inequality**.
9. Sets and dictionaries have special membership behavior; for a dictionary, `in` checks keys by default.
10. Bitwise operators work on the binary representation of integers.
11. Parentheses can be used to explicitly control evaluation order.
12. A useful precedence sequence to remember is:

   `()` → `**` → unary operators → `* / // %` → `+ -` → comparisons → `not` → `and` → `or`
